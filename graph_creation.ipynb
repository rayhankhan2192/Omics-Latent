```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches

# --- Data Preparation ---
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.00, 0.00, 0.00, 0.51, 0.00],
    'Recall':    [0.00, 0.00, 0.00, 0.90, 0.00],
    'F1-score':  [0.00, 0.00, 0.00, 0.67, 0.00],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.36, 0.51],
    'Recall':    [0.08, 0.87],
    'F1-score':  [0.13, 0.64],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Normalize support for Marimekko chart
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# --- Figure Layout ---
fig = plt.figure(figsize=(16, 10))
gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})
colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
colors_rosmap = ['#E67E22', '#1ABC9C']

# ==========================================
# PANEL A: Parallel Coordinates Plot
# ==========================================
gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
ax_a1 = fig.add_subplot(gs_a[0])
ax_a2 = fig.add_subplot(gs_a[1])

def plot_parallel_coords(ax, df, colors, title):
    metrics = ['Precision', 'Recall', 'F1-score']
    x = range(len(metrics))
    for i, row in df.iterrows():
        y = row[metrics].values
        ax.plot(x, y, marker='o', linewidth=2, label=f"Class {row['Class']}", color=colors[i % len(colors)], alpha=0.8)
    ax.set_xticks(x)
    ax.set_xticklabels(metrics)
    ax.set_ylim(-0.05, 1.05)
    ax.set_title(title, fontweight='bold', fontsize=10)
    ax.grid(True, axis='y', linestyle='--', alpha=0.5)
    if title.startswith("ROSMAP"):
        ax.set_yticklabels([])
        ax.legend(loc='upper right', fontsize=8)
    else:
        ax.set_ylabel("Score")
        ax.legend(loc='upper left', fontsize=8)

plot_parallel_coords(ax_a1, df_brca, colors_brca, "BRCA: Class Profiles")
plot_parallel_coords(ax_a2, df_rosmap, colors_rosmap, "ROSMAP: Class Profiles")
ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

# ==========================================
# PANEL B: Lollipop Chart
# ==========================================
ax_b = fig.add_subplot(gs[0, 1])
df_brca['Dataset'] = 'BRCA'
df_rosmap['Dataset'] = 'ROSMAP'
combined = pd.concat([df_brca, df_rosmap]).reset_index(drop=True)
combined['Label'] = combined['Dataset'] + " Class " + combined['Class']
combined = combined.sort_values('F1-score', ascending=True)

y_range = range(len(combined))
ax_b.hlines(y=y_range, xmin=0, xmax=combined['F1-score'], color='gray', alpha=0.5, linewidth=1)
for i, (val, label) in enumerate(zip(combined['F1-score'], combined['Label'])):
    col = '#E74C3C' if 'BRCA' in label else '#3498DB'
    ax_b.scatter(val, i, color=col, s=100, zorder=4)
    ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')

ax_b.set_yticks(y_range)
ax_b.set_yticklabels(combined['Label'])
ax_b.set_xlabel("F1-Score")
ax_b.set_title("Ranked Class Performance (Lollipop Chart)", fontweight='bold', fontsize=10)
ax_b.set_xlim(0, 1.1)
ax_b.grid(axis='x', linestyle='--', alpha=0.5)
ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

# ==========================================
# PANEL C: Marimekko Chart
# ==========================================
gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
ax_c1 = fig.add_subplot(gs_c[0])
ax_c2 = fig.add_subplot(gs_c[1])

def plot_marimekko(ax, df, colors, title):
    current_x = 0
    for i, row in df.iterrows():
        width = row['Support_Pct']
        height = row['F1-score']
        rect = patches.Rectangle((current_x, 0), width, height, linewidth=1, edgecolor='white', facecolor=colors[i % len(colors)], alpha=0.8)
        ax.add_patch(rect)
        if width > 0.1:
            ax.text(current_x + width/2, height/2, f"Cls {row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
        current_x += width
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1.05)
    ax.set_xlabel("Cumulative Support")
    ax.set_title(title, fontweight='bold', fontsize=10)
    if title.startswith("ROSMAP"):
        ax.set_yticklabels([])
    else:
        ax.set_ylabel("F1-Score")

plot_marimekko(ax_c1, df_brca, colors_brca, "BRCA: Impact Map\n(Width=Support, Height=F1)")
plot_marimekko(ax_c2, df_rosmap, colors_rosmap, "ROSMAP: Impact Map")
ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

# ==========================================
# PANEL D: Slope Graph
# ==========================================
gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
ax_d1 = fig.add_subplot(gs_d[0])
ax_d2 = fig.add_subplot(gs_d[1])

def plot_slope(ax, df, colors, title):
    for i, row in df.iterrows():
        ax.plot([0, 1], [row['Precision'], row['Recall']], color=colors[i % len(colors)], alpha=0.8, linewidth=2, marker='o')
    ax.set_xlim(-0.2, 1.2)
    ax.set_ylim(-0.05, 1.05)
    ax.set_xticks([0, 1])
    ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
    ax.set_title(title, fontweight='bold', fontsize=10)
    ax.grid(axis='y', linestyle=':', alpha=0.5)
    ax.spines['left'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['top'].set_visible(False)
    if title.startswith("ROSMAP"):
        ax.set_yticklabels([])
    else:
        ax.set_ylabel("Score")

plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

plt.tight_layout()
plt.subplots_adjust(wspace=0.2, hspace=0.3)
plt.savefig('publication_figure_v2.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from math import pi
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ==========================================
# 1. SHARED DATA PREPARATION
# ==========================================
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.00, 0.00, 0.00, 0.51, 0.00],
    'Recall':    [0.00, 0.00, 0.00, 0.90, 0.00],
    'F1-score':  [0.00, 0.00, 0.00, 0.67, 0.00],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.36, 0.51],
    'Recall':    [0.08, 0.87],
    'F1-score':  [0.13, 0.64],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Normalize support for Marimekko chart (Figure 2)
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# Global Font Settings
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})


# ==========================================
# 2. GENERATE FIGURE 1 (Overview)
# ==========================================
def generate_figure_1():
    print("Generating Figure 1: Overview...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Bar Charts ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.3)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_bar(ax, df, title, pal):
        x = np.arange(len(df['Class']))
        width = 0.25
        metrics = ['Precision', 'Recall', 'F1-score']
        for i, metric in enumerate(metrics):
            ax.bar(x + i*width, df[metric], width=width, label=metric, color=pal[i], edgecolor='white')
        ax.set_xticks(x + width)
        ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')
        ax.set_ylim(0, 1.05)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        if "ROSMAP" in title: ax.legend(fontsize=8, loc='upper right')

    colors = ['#5DADE2', '#58D68D', '#EC7063']
    plot_bar(ax_a1, df_brca, "BRCA Performance", colors)
    plot_bar(ax_a2, df_rosmap, "ROSMAP Performance", colors)
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Heatmaps ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.3)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    sns.heatmap(df_brca.set_index('Class')[['Precision', 'Recall', 'F1-score']], 
                annot=True, cmap='Reds', fmt='.2f', cbar=False, ax=ax_b1, vmin=0, vmax=1)
    ax_b1.set_title("BRCA Heatmap", fontsize=10, fontweight='bold')
    sns.heatmap(df_rosmap.set_index('Class')[['Precision', 'Recall', 'F1-score']], 
                annot=True, cmap='Blues', fmt='.2f', cbar=False, ax=ax_b2, vmin=0, vmax=1)
    ax_b2.set_title("ROSMAP Heatmap", fontsize=10, fontweight='bold')
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Radar Charts ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.3)
    ax_c1 = fig.add_subplot(gs_c[0], polar=True)
    ax_c2 = fig.add_subplot(gs_c[1], polar=True)

    def plot_radar(ax, df, title):
        categories = ['Precision', 'Recall', 'F1-score']
        N = len(categories)
        angles = [n / float(N) * 2 * pi for n in range(N)]
        angles += angles[:1]
        ax.set_theta_offset(pi / 2)
        ax.set_theta_direction(-1)
        ax.set_xticks(angles[:-1])
        ax.set_xticklabels(categories, fontsize=8)
        ax.set_yticks([0.2, 0.4, 0.6, 0.8])
        ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=6)
        ax.set_ylim(0, 1)
        for index, row in df.iterrows():
            values = row[categories].values.flatten().tolist()
            values += values[:1]
            ax.plot(angles, values, linewidth=1.5, linestyle='solid', label=f"Class {row['Class']}")
            ax.fill(angles, values, alpha=0.05)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=15)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fontsize=8, ncol=2)

    plot_radar(ax_c1, df_brca, "BRCA Profile")
    plot_radar(ax_c2, df_rosmap, "ROSMAP Profile")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Bubble Chart ---
    ax_d = fig.add_subplot(gs[1, 1])
    ax_d.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*10, alpha=0.6, c='#E74C3C', label='BRCA', edgecolors='black')
    ax_d.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*10, alpha=0.6, c='#3498DB', label='ROSMAP', edgecolors='black')
    
    for i, txt in enumerate(df_brca['Class']):
        ax_d.annotate(f"B-{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    for i, txt in enumerate(df_rosmap['Class']):
        ax_d.annotate(f"R-{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')

    ax_d.set_xlabel('Recall')
    ax_d.set_ylabel('Precision')
    ax_d.set_title("Precision vs Recall (Size = Support)", fontsize=10, fontweight='bold')
    ax_d.grid(True, linestyle='--', alpha=0.5)
    ax_d.legend(loc='upper left', fontsize=9)
    ax_d.text(-0.1, 1.05, "D", transform=ax_d.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.25, hspace=0.3)
    plt.savefig('figure_1_overview.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 3. GENERATE FIGURE 2 (Analytical)
# ==========================================
def generate_figure_2():
    print("Generating Figure 2: Analytical...")
    fig = plt.figure(figsize=(16, 10))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
    colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
    colors_rosmap = ['#E67E22', '#1ABC9C']

    # --- Panel A: Parallel Coords ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_pc(ax, df, cols, title):
        metrics = ['Precision', 'Recall', 'F1-score']
        x = range(len(metrics))
        for i, row in df.iterrows():
            ax.plot(x, row[metrics].values, marker='o', linewidth=2, label=f"Cls {row['Class']}", color=cols[i%len(cols)], alpha=0.8)
        ax.set_xticks(x)
        ax.set_xticklabels(metrics)
        ax.set_ylim(-0.05, 1.05)
        ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle='--', alpha=0.5)
        if "ROSMAP" in title:
            ax.set_yticklabels([])
            ax.legend(loc='upper right', fontsize=8)
        else:
            ax.set_ylabel("Score")
            ax.legend(loc='upper left', fontsize=8)

    plot_pc(ax_a1, df_brca, colors_brca, "BRCA Profiles")
    plot_pc(ax_a2, df_rosmap, colors_rosmap, "ROSMAP Profiles")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Lollipop Chart ---
    ax_b = fig.add_subplot(gs[0, 1])
    df_temp = pd.concat([df_brca.assign(D='BRCA'), df_rosmap.assign(D='ROSMAP')]).reset_index(drop=True)
    df_temp['Label'] = df_temp['D'] + " Cls " + df_temp['Class']
    df_temp = df_temp.sort_values('F1-score')
    
    y = range(len(df_temp))
    ax_b.hlines(y=y, xmin=0, xmax=df_temp['F1-score'], color='gray', alpha=0.5)
    for i, (val, label) in enumerate(zip(df_temp['F1-score'], df_temp['Label'])):
        col = '#E74C3C' if 'BRCA' in label else '#3498DB'
        ax_b.scatter(val, i, color=col, s=100, zorder=4)
        ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')
    
    ax_b.set_yticks(y)
    ax_b.set_yticklabels(df_temp['Label'])
    ax_b.set_title("Ranked Class Performance", fontweight='bold', fontsize=10)
    ax_b.set_xlabel("F1-Score")
    ax_b.set_xlim(0, 1.1)
    ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Marimekko ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])

    def plot_mari(ax, df, cols, title):
        curr_x = 0
        for i, row in df.iterrows():
            w, h = row['Support_Pct'], row['F1-score']
            rect = patches.Rectangle((curr_x, 0), w, h, linewidth=1, edgecolor='white', facecolor=cols[i%len(cols)], alpha=0.8)
            ax.add_patch(rect)
            if w > 0.1: ax.text(curr_x + w/2, h/2, f"Cls {row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
            curr_x += w
        ax.set_xlim(0, 1); ax.set_ylim(0, 1.05)
        ax.set_xlabel("Cumulative Support")
        ax.set_title(title, fontweight='bold', fontsize=10)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("F1-Score")

    plot_mari(ax_c1, df_brca, colors_brca, "BRCA Impact Map")
    plot_mari(ax_c2, df_rosmap, colors_rosmap, "ROSMAP Impact Map")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Slope Graph ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_slope(ax, df, cols, title):
        for i, row in df.iterrows():
            ax.plot([0, 1], [row['Precision'], row['Recall']], color=cols[i%len(cols)], alpha=0.8, linewidth=2, marker='o')
        ax.set_xticks([0, 1]); ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
        ax.set_ylim(-0.05, 1.05); ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        ax.spines['left'].set_visible(False); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("Score")

    plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
    plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
    ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.2, hspace=0.3)
    plt.savefig('figure_2_analytical.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 4. GENERATE FIGURE 3 (Comparative)
# ==========================================
def generate_figure_3():
    print("Generating Figure 3: Comparative...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Butterfly ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_butterfly(ax, df, title):
        y = np.arange(len(df['Class']))
        ax.barh(y, -df['Precision'], color='#5DADE2', label='Precision', height=0.6)
        ax.barh(y, df['Recall'], color='#58D68D', label='Recall', height=0.6)
        ax.set_yticks(y); ax.set_yticklabels([f"Cls {c}" for c in df['Class']])
        ax.axvline(0, color='black', linewidth=0.8)
        ticks = np.linspace(-1, 1, 5)
        ax.set_xticks(ticks); ax.set_xticklabels([f"{abs(t):.1f}" for t in ticks], fontsize=8)
        ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=9)

    plot_butterfly(ax_a1, df_brca, "BRCA Metric Divergence")
    plot_butterfly(ax_a2, df_rosmap, "ROSMAP Metric Divergence")
    ax_a1.text(-0.2, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Donut ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.1)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_donut(ax, df, title):
        norm = mcolors.Normalize(vmin=0, vmax=0.8)
        cmap = cm.get_cmap('RdYlGn')
        colors = [cmap(norm(val)) for val in df['F1-score']]
        wedges, _ = ax.pie(df['Support'], colors=colors, startangle=90, wedgeprops=dict(width=0.4, edgecolor='white'))
        avg = np.average(df['F1-score'], weights=df['Support'])
        ax.text(0, 0, f"Avg F1\n{avg:.2f}", ha='center', va='center', fontweight='bold', fontsize=10)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
        ax.legend(wedges, [f"C{c}" for c in df['Class']], title="Classes", loc="center left", bbox_to_anchor=(0.85, 0, 0.5, 1), fontsize=8)

    plot_donut(ax_b1, df_brca, "BRCA (Color=F1)")
    plot_donut(ax_b2, df_rosmap, "ROSMAP (Color=F1)")
    ax_b1.text(-0.1, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Dumbbell ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.25)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])

    def plot_dumb(ax, df, title):
        for i, row in df.iterrows():
            ax.plot([row['Precision'], row['Recall']], [i, i], color='gray', zorder=1)
            ax.scatter(row['Precision'], i, color='#5DADE2', zorder=2, s=60)
            ax.scatter(row['Recall'], i, color='#58D68D', zorder=2, s=60)
        ax.set_yticks(range(len(df))); ax.set_yticklabels(df['Class'])
        ax.set_xlim(-0.1, 1.1); ax.set_title(title, fontsize=10, fontweight='bold')
        ax.grid(axis='x', linestyle='--', alpha=0.5)

    plot_dumb(ax_c1, df_brca, "BRCA Gap Analysis")
    plot_dumb(ax_c2, df_rosmap, "ROSMAP Gap Analysis")
    ax_c1.text(-0.2, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Dual Axis ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.3)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_dual(ax, df, title):
        x = np.arange(len(df['Class']))
        ax.bar(x, df['Support'], color='lightgray', alpha=0.8, width=0.6)
        ax.set_ylabel('Support', color='gray', fontsize=8)
        ax2 = ax.twinx()
        ax2.plot(x, df['F1-score'], color='#E74C3C', marker='o', linewidth=2)
        ax2.set_ylabel('F1-Score', color='#E74C3C', fontsize=8)
        ax.set_xticks(x); ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dual(ax_d1, df_brca, "BRCA: Vol vs Qual")
    plot_dual(ax_d2, df_rosmap, "ROSMAP: Vol vs Qual")
    ax_d1.text(-0.25, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.25, hspace=0.3)
    plt.savefig('figure_3_comparative.png', dpi=300, bbox_inches='tight')
    plt.close()

# ==========================================
# 5. EXECUTE ALL
# ==========================================
if __name__ == "__main__":
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    print("All figures generated successfully.")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches

# --- Data Preparation ---
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.00, 0.00, 0.00, 0.51, 0.00],
    'Recall':    [0.00, 0.00, 0.00, 0.90, 0.00],
    'F1-score':  [0.00, 0.00, 0.00, 0.67, 0.00],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.36, 0.51],
    'Recall':    [0.08, 0.87],
    'F1-score':  [0.13, 0.64],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Calculate Weighted Averages for Reference
brca_w_avg = np.average(df_brca['F1-score'], weights=df_brca['Support'])
rosmap_w_avg = np.average(df_rosmap['F1-score'], weights=df_rosmap['Support'])

# --- Figure Layout ---
fig = plt.figure(figsize=(16, 12))
gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})

# ==========================================
# PANEL A: Nightingale Rose (Coxcomb) Chart
# ==========================================
# Polar bar charts.
gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
ax_a1 = fig.add_subplot(gs_a[0], polar=True)
ax_a2 = fig.add_subplot(gs_a[1], polar=True)

def plot_rose(ax, df, title):
    N = len(df)
    theta = np.linspace(0.0, 2 * np.pi, N, endpoint=False)
    width = (2*np.pi) / N
    
    # We plot F1 score as the radius
    radii = df['F1-score']
    colors = plt.cm.viridis(radii / 1.0) # Color by score magnitude
    
    bars = ax.bar(theta, radii, width=width, bottom=0.0, color=colors, alpha=0.8, edgecolor='white')
    
    # Labeling
    ax.set_xticks(theta)
    ax.set_xticklabels([f"C{c}" for c in df['Class']], fontsize=9, fontweight='bold')
    ax.set_yticklabels([]) # Hide radial labels
    ax.set_ylim(0, 1)
    ax.set_title(title, fontsize=10, fontweight='bold', pad=20)

plot_rose(ax_a1, df_brca, "BRCA: Performance Profile")
plot_rose(ax_a2, df_rosmap, "ROSMAP: Performance Profile")
ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

# ==========================================
# PANEL B: Diverging Deviation Plot
# ==========================================
# How far is each class from the Weighted Average?
gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.25)
ax_b1 = fig.add_subplot(gs_b[0])
ax_b2 = fig.add_subplot(gs_b[1])

def plot_deviation(ax, df, avg_score, title):
    x = np.arange(len(df['Class']))
    deviation = df['F1-score'] - avg_score
    
    # Color bars based on direction (Red=Below Avg, Green=Above Avg)
    colors = ['#E74C3C' if d < 0 else '#2ECC71' for d in deviation]
    
    ax.bar(x, deviation, color=colors, alpha=0.8)
    ax.axhline(0, color='black', linewidth=1)
    
    # Add text labels
    for i, v in enumerate(deviation):
        offset = 0.05 if v >= 0 else -0.10
        ax.text(i, v + (0.02 if v >=0 else -0.05), f"{v:+.2f}", ha='center', fontsize=8, fontweight='bold')
    
    ax.set_xticks(x)
    ax.set_xticklabels(df['Class'])
    ax.set_ylim(-0.8, 0.8)
    ax.set_title(title, fontsize=10, fontweight='bold')
    
    # Annotate the zero line
    ax.text(len(df)-1, 0.02, f"Avg: {avg_score:.2f}", ha='right', va='bottom', fontsize=8, fontstyle='italic', color='gray')

plot_deviation(ax_b1, df_brca, brca_w_avg, "BRCA: Deviation from Mean")
plot_deviation(ax_b2, df_rosmap, rosmap_w_avg, "ROSMAP: Deviation from Mean")
ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

# ==========================================
# PANEL C: "Assessment Zones" (Precision vs Recall with Zones)
# ==========================================
ax_c = fig.add_subplot(gs[1, 0])

# Draw Background Zones
# Poor (Red) < 0.5, Fair (Yellow) 0.5-0.75, Good (Green) > 0.75
# Since it's 2D, we can draw rects or just color based on x+y or simple boxes.
# Let's use simple boxes for simplicity: High Precision & High Recall is best.
rect_poor = patches.Rectangle((0, 0), 1, 1, color='#FDEDEC') # Light Red base
rect_good = patches.Rectangle((0.6, 0.6), 0.4, 0.4, color='#EAFAF1') # Light Green top right

ax_c.add_patch(rect_poor)
ax_c.add_patch(rect_good)

# Plot Points
ax_c.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*8, c='#3498DB', label='BRCA', edgecolors='black', alpha=0.8)
ax_c.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*8, c='#E67E22', label='ROSMAP', edgecolors='black', alpha=0.8)

# Labels
for i, txt in enumerate(df_brca['Class']):
    ax_c.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, -10), textcoords='offset points', ha='center', fontsize=8)
for i, txt in enumerate(df_rosmap['Class']):
    ax_c.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', ha='center', fontsize=8)

ax_c.set_xlim(0, 1.05)
ax_c.set_ylim(0, 1.05)
ax_c.set_xlabel("Recall")
ax_c.set_ylabel("Precision")
ax_c.set_title("Performance Assessment Zones\n(Green = Target Area)", fontweight='bold', fontsize=10)
ax_c.grid(True, linestyle=':', alpha=0.5)
ax_c.legend(loc='lower left')
ax_c.text(-0.1, 1.1, "C", transform=ax_c.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

# ==========================================
# PANEL D: Estimated Outcome Stack (Correct vs Error)
# ==========================================
# TP = Recall * Support
# Error = Support - TP
gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.25)
ax_d1 = fig.add_subplot(gs_d[0])
ax_d2 = fig.add_subplot(gs_d[1])

def plot_outcome(ax, df, title):
    x = np.arange(len(df['Class']))
    
    # Calculate counts
    correct = df['Recall'] * df['Support']
    error = df['Support'] - correct
    
    # Plot Stacked Bars
    ax.bar(x, correct, color='#2ECC71', label='Correct (TP)', width=0.6)
    ax.bar(x, error, bottom=correct, color='#E74C3C', label='Missed/Wrong', width=0.6)
    
    ax.set_xticks(x)
    ax.set_xticklabels(df['Class'])
    ax.set_title(title, fontsize=10, fontweight='bold')
    
    if title.startswith("BRCA"):
        ax.set_ylabel("Number of Samples")
        ax.legend(loc='upper right', fontsize=8)

plot_outcome(ax_d1, df_brca, "BRCA: Sample Outcomes")
plot_outcome(ax_d2, df_rosmap, "ROSMAP: Sample Outcomes")
ax_d1.text(-0.2, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

plt.tight_layout()
plt.subplots_adjust(wspace=0.2, hspace=0.3)
plt.savefig('figure_4_diagnostic.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from math import pi
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ==========================================
# 0. DATA PREPARATION (SHARED)
# ==========================================
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.00, 0.00, 0.00, 0.51, 0.00],
    'Recall':    [0.00, 0.00, 0.00, 0.90, 0.00],
    'F1-score':  [0.00, 0.00, 0.00, 0.67, 0.00],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.36, 0.51],
    'Recall':    [0.08, 0.87],
    'F1-score':  [0.13, 0.64],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Calculate percentages for Marimekko (Fig 2)
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# Calculate weighted averages for Deviation Plot (Fig 4)
brca_w_avg = np.average(df_brca['F1-score'], weights=df_brca['Support'])
rosmap_w_avg = np.average(df_rosmap['F1-score'], weights=df_rosmap['Support'])

# Global Plot Settings
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})


# ==========================================
# 1. FIGURE 1: OVERVIEW (Bar, Heatmap, Radar, Bubble)
# ==========================================
def generate_figure_1():
    print("Generating Figure 1: Overview...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Bar Charts ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.3)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    colors = ['#5DADE2', '#58D68D', '#EC7063']

    def plot_bar(ax, df, title):
        x = np.arange(len(df['Class']))
        width = 0.25
        metrics = ['Precision', 'Recall', 'F1-score']
        for i, metric in enumerate(metrics):
            ax.bar(x + i*width, df[metric], width=width, label=metric, color=colors[i], edgecolor='white')
        ax.set_xticks(x + width)
        ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')
        ax.set_ylim(0, 1.05)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        if "ROSMAP" in title: ax.legend(fontsize=8, loc='upper right')

    plot_bar(ax_a1, df_brca, "BRCA Performance")
    plot_bar(ax_a2, df_rosmap, "ROSMAP Performance")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Heatmaps ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.3)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])
    
    sns.heatmap(df_brca.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Reds', fmt='.2f', cbar=False, ax=ax_b1, vmin=0, vmax=1)
    ax_b1.set_title("BRCA Heatmap", fontsize=10, fontweight='bold')
    sns.heatmap(df_rosmap.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Blues', fmt='.2f', cbar=False, ax=ax_b2, vmin=0, vmax=1)
    ax_b2.set_title("ROSMAP Heatmap", fontsize=10, fontweight='bold')
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Radar Charts ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.3)
    ax_c1 = fig.add_subplot(gs_c[0], polar=True)
    ax_c2 = fig.add_subplot(gs_c[1], polar=True)

    def plot_radar(ax, df, title):
        categories = ['Precision', 'Recall', 'F1-score']
        N = len(categories)
        angles = [n / float(N) * 2 * pi for n in range(N)]
        angles += angles[:1]
        ax.set_theta_offset(pi / 2); ax.set_theta_direction(-1)
        ax.set_xticks(angles[:-1]); ax.set_xticklabels(categories, fontsize=8)
        ax.set_ylim(0, 1); ax.set_yticks([0.2, 0.4, 0.6, 0.8]); ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=6)
        for index, row in df.iterrows():
            values = row[categories].values.flatten().tolist()
            values += values[:1]
            ax.plot(angles, values, linewidth=1.5, linestyle='solid', label=f"Cls {row['Class']}")
            ax.fill(angles, values, alpha=0.05)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=15)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fontsize=8, ncol=2)

    plot_radar(ax_c1, df_brca, "BRCA Profile")
    plot_radar(ax_c2, df_rosmap, "ROSMAP Profile")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Bubble Chart ---
    ax_d = fig.add_subplot(gs[1, 1])
    ax_d.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*10, alpha=0.6, c='#E74C3C', label='BRCA', edgecolors='black')
    ax_d.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*10, alpha=0.6, c='#3498DB', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_d.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    for i, txt in enumerate(df_rosmap['Class']):
        ax_d.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    ax_d.set_xlabel('Recall'); ax_d.set_ylabel('Precision'); ax_d.set_title("Precision vs Recall (Size = Support)", fontsize=10, fontweight='bold')
    ax_d.grid(True, linestyle='--', alpha=0.5); ax_d.legend(loc='upper left', fontsize=9)
    ax_d.text(-0.1, 1.05, "D", transform=ax_d.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.tight_layout()
    plt.savefig('figure_1_overview.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 2. FIGURE 2: ANALYTICAL (Parallel, Lollipop, Marimekko, Slope)
# ==========================================
def generate_figure_2():
    print("Generating Figure 2: Analytical...")
    fig = plt.figure(figsize=(16, 10))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
    colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
    colors_rosmap = ['#E67E22', '#1ABC9C']

    # --- Panel A: Parallel Coords ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    
    def plot_pc(ax, df, cols, title):
        metrics = ['Precision', 'Recall', 'F1-score']
        x = range(len(metrics))
        for i, row in df.iterrows():
            ax.plot(x, row[metrics].values, marker='o', linewidth=2, label=f"Cls {row['Class']}", color=cols[i%len(cols)], alpha=0.8)
        ax.set_xticks(x); ax.set_xticklabels(metrics); ax.set_ylim(-0.05, 1.05)
        ax.set_title(title, fontweight='bold', fontsize=10); ax.grid(axis='y', linestyle='--', alpha=0.5)
        if "ROSMAP" in title: ax.set_yticklabels([]); ax.legend(loc='upper right', fontsize=8)
        else: ax.set_ylabel("Score"); ax.legend(loc='upper left', fontsize=8)

    plot_pc(ax_a1, df_brca, colors_brca, "BRCA Profiles")
    plot_pc(ax_a2, df_rosmap, colors_rosmap, "ROSMAP Profiles")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Lollipop Chart ---
    ax_b = fig.add_subplot(gs[0, 1])
    df_temp = pd.concat([df_brca.assign(D='BRCA'), df_rosmap.assign(D='ROSMAP')]).sort_values('F1-score')
    df_temp['Label'] = df_temp['D'] + " Cls " + df_temp['Class']
    y = range(len(df_temp))
    ax_b.hlines(y=y, xmin=0, xmax=df_temp['F1-score'], color='gray', alpha=0.5)
    for i, (val, label) in enumerate(zip(df_temp['F1-score'], df_temp['Label'])):
        col = '#E74C3C' if 'BRCA' in label else '#3498DB'
        ax_b.scatter(val, i, color=col, s=100, zorder=4)
        ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')
    ax_b.set_yticks(y); ax_b.set_yticklabels(df_temp['Label'])
    ax_b.set_title("Ranked Class Performance", fontweight='bold', fontsize=10); ax_b.set_xlabel("F1-Score")
    ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Marimekko ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_mari(ax, df, cols, title):
        curr_x = 0
        for i, row in df.iterrows():
            w, h = row['Support_Pct'], row['F1-score']
            rect = patches.Rectangle((curr_x, 0), w, h, linewidth=1, edgecolor='white', facecolor=cols[i%len(cols)], alpha=0.8)
            ax.add_patch(rect)
            if w > 0.1: ax.text(curr_x + w/2, h/2, f"C{row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
            curr_x += w
        ax.set_xlim(0, 1); ax.set_ylim(0, 1.05); ax.set_title(title, fontweight='bold', fontsize=10); ax.set_xlabel("Cumulative Support")
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("F1-Score")

    plot_mari(ax_c1, df_brca, colors_brca, "BRCA Impact Map")
    plot_mari(ax_c2, df_rosmap, colors_rosmap, "ROSMAP Impact Map")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Slope Graph ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])
    
    def plot_slope(ax, df, cols, title):
        for i, row in df.iterrows():
            ax.plot([0, 1], [row['Precision'], row['Recall']], color=cols[i%len(cols)], alpha=0.8, linewidth=2, marker='o')
        ax.set_xticks([0, 1]); ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
        ax.set_ylim(-0.05, 1.05); ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        ax.spines['left'].set_visible(False); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("Score")

    plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
    plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
    ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.tight_layout()
    plt.savefig('figure_2_analytical.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 3. FIGURE 3: COMPARATIVE (Butterfly, Donut, Dumbbell, Dual-Axis)
# ==========================================
def generate_figure_3():
    print("Generating Figure 3: Comparative...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Butterfly ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_butterfly(ax, df, title):
        y = np.arange(len(df['Class']))
        ax.barh(y, -df['Precision'], color='#5DADE2', label='Precision', height=0.6)
        ax.barh(y, df['Recall'], color='#58D68D', label='Recall', height=0.6)
        ax.set_yticks(y); ax.set_yticklabels([f"Cls {c}" for c in df['Class']])
        ax.axvline(0, color='black', linewidth=0.8)
        ticks = np.linspace(-1, 1, 5)
        ax.set_xticks(ticks); ax.set_xticklabels([f"{abs(t):.1f}" for t in ticks], fontsize=8)
        ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=9)

    plot_butterfly(ax_a1, df_brca, "BRCA Metric Divergence")
    plot_butterfly(ax_a2, df_rosmap, "ROSMAP Metric Divergence")
    ax_a1.text(-0.2, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Donut ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.1)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_donut(ax, df, title):
        norm = mcolors.Normalize(vmin=0, vmax=0.8)
        cmap = cm.get_cmap('RdYlGn')
        colors = [cmap(norm(val)) for val in df['F1-score']]
        wedges, _ = ax.pie(df['Support'], colors=colors, startangle=90, wedgeprops=dict(width=0.4, edgecolor='white'))
        avg = np.average(df['F1-score'], weights=df['Support'])
        ax.text(0, 0, f"Avg F1\n{avg:.2f}", ha='center', va='center', fontweight='bold', fontsize=10)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
        ax.legend(wedges, [f"C{c}" for c in df['Class']], title="Classes", loc="center left", bbox_to_anchor=(0.85, 0, 0.5, 1), fontsize=8)

    plot_donut(ax_b1, df_brca, "BRCA (Color=F1)")
    plot_donut(ax_b2, df_rosmap, "ROSMAP (Color=F1)")
    ax_b1.text(-0.1, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Dumbbell ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.25)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_dumb(ax, df, title):
        for i, row in df.iterrows():
            ax.plot([row['Precision'], row['Recall']], [i, i], color='gray', zorder=1)
            ax.scatter(row['Precision'], i, color='#5DADE2', zorder=2, s=60)
            ax.scatter(row['Recall'], i, color='#58D68D', zorder=2, s=60)
        ax.set_yticks(range(len(df))); ax.set_yticklabels(df['Class'])
        ax.set_xlim(-0.1, 1.1); ax.set_title(title, fontsize=10, fontweight='bold'); ax.grid(axis='x', linestyle='--', alpha=0.5)

    plot_dumb(ax_c1, df_brca, "BRCA Gap Analysis")
    plot_dumb(ax_c2, df_rosmap, "ROSMAP Gap Analysis")
    ax_c1.text(-0.2, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Dual Axis ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.3)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_dual(ax, df, title):
        x = np.arange(len(df['Class']))
        ax.bar(x, df['Support'], color='lightgray', alpha=0.8, width=0.6)
        ax.set_ylabel('Support', color='gray', fontsize=8)
        ax2 = ax.twinx()
        ax2.plot(x, df['F1-score'], color='#E74C3C', marker='o', linewidth=2)
        ax2.set_ylabel('F1-Score', color='#E74C3C', fontsize=8)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dual(ax_d1, df_brca, "BRCA: Vol vs Qual")
    plot_dual(ax_d2, df_rosmap, "ROSMAP: Vol vs Qual")
    ax_d1.text(-0.25, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.tight_layout()
    plt.savefig('figure_3_comparative.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 4. FIGURE 4: DIAGNOSTIC (Rose, Deviation, Zones, Outcome)
# ==========================================
def generate_figure_4():
    print("Generating Figure 4: Diagnostic...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Rose ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0], polar=True)
    ax_a2 = fig.add_subplot(gs_a[1], polar=True)
    
    def plot_rose(ax, df, title):
        N = len(df); theta = np.linspace(0.0, 2 * np.pi, N, endpoint=False); width = (2*np.pi) / N
        radii = df['F1-score']; colors = plt.cm.viridis(radii / 1.0)
        ax.bar(theta, radii, width=width, bottom=0.0, color=colors, alpha=0.8, edgecolor='white')
        ax.set_xticks(theta); ax.set_xticklabels([f"C{c}" for c in df['Class']], fontsize=9, fontweight='bold')
        ax.set_yticklabels([]); ax.set_ylim(0, 1); ax.set_title(title, fontsize=10, fontweight='bold', pad=20)

    plot_rose(ax_a1, df_brca, "BRCA: Performance Profile")
    plot_rose(ax_a2, df_rosmap, "ROSMAP: Performance Profile")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Deviation ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.25)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_dev(ax, df, avg, title):
        x = np.arange(len(df)); dev = df['F1-score'] - avg
        cols = ['#E74C3C' if d < 0 else '#2ECC71' for d in dev]
        ax.bar(x, dev, color=cols, alpha=0.8); ax.axhline(0, color='black', linewidth=1)
        for i, v in enumerate(dev):
            ax.text(i, v + (0.02 if v >=0 else -0.05), f"{v:+.2f}", ha='center', fontsize=8, fontweight='bold')
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_ylim(-0.8, 0.8); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dev(ax_b1, df_brca, brca_w_avg, "BRCA: Deviation")
    plot_dev(ax_b2, df_rosmap, rosmap_w_avg, "ROSMAP: Deviation")
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Zones ---
    ax_c = fig.add_subplot(gs[1, 0])
    ax_c.add_patch(patches.Rectangle((0, 0), 1, 1, color='#FDEDEC'))
    ax_c.add_patch(patches.Rectangle((0.6, 0.6), 0.4, 0.4, color='#EAFAF1'))
    ax_c.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*8, c='#3498DB', label='BRCA', edgecolors='black')
    ax_c.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*8, c='#E67E22', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_c.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, -10), textcoords='offset points', ha='center', fontsize=8)
    for i, txt in enumerate(df_rosmap['Class']):
        ax_c.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', ha='center', fontsize=8)
    ax_c.set_xlim(0, 1.05); ax_c.set_ylim(0, 1.05); ax_c.set_xlabel("Recall"); ax_c.set_ylabel("Precision")
    ax_c.set_title("Performance Assessment Zones", fontweight='bold', fontsize=10); ax_c.legend(loc='lower left')
    ax_c.text(-0.1, 1.1, "C", transform=ax_c.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Outcomes ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.25)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_out(ax, df, title):
        x = np.arange(len(df)); corr = df['Recall'] * df['Support']; err = df['Support'] - corr
        ax.bar(x, corr, color='#2ECC71', label='Correct', width=0.6)
        ax.bar(x, err, bottom=corr, color='#E74C3C', label='Error', width=0.6)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper right', fontsize=8)

    plot_out(ax_d1, df_brca, "BRCA Outcomes")
    plot_out(ax_d2, df_rosmap, "ROSMAP Outcomes")
    ax_d1.text(-0.2, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.tight_layout()
    plt.savefig('figure_4_diagnostic.png', dpi=300, bbox_inches='tight')
    plt.close()

# ==========================================
# 5. EXECUTION
# ==========================================
if __name__ == "__main__":
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    generate_figure_4()
    print("All 4 figures generated successfully!")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from math import pi
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ==========================================
# 0. DATA PREPARATION (SHARED)
# ==========================================
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.00, 0.00, 0.00, 0.51, 0.00],
    'Recall':    [0.00, 0.00, 0.00, 0.90, 0.00],
    'F1-score':  [0.00, 0.00, 0.00, 0.67, 0.00],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.36, 0.51],
    'Recall':    [0.08, 0.87],
    'F1-score':  [0.13, 0.64],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Calculate percentages for Marimekko (Fig 2)
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# Calculate weighted averages for Deviation Plot (Fig 4)
brca_w_avg = np.average(df_brca['F1-score'], weights=df_brca['Support'])
rosmap_w_avg = np.average(df_rosmap['F1-score'], weights=df_rosmap['Support'])

# Global Plot Settings
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})
MAIN_TITLE = "View-Specific Stacked Generalization Ensemble"


# ==========================================
# 1. FIGURE 1: OVERVIEW (Bar, Heatmap, Radar, Bubble)
# ==========================================
def generate_figure_1():
    print("Generating Figure 1: Overview...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Bar Charts ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.3)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    colors = ['#5DADE2', '#58D68D', '#EC7063']

    def plot_bar(ax, df, title):
        x = np.arange(len(df['Class']))
        width = 0.25
        metrics = ['Precision', 'Recall', 'F1-score']
        for i, metric in enumerate(metrics):
            ax.bar(x + i*width, df[metric], width=width, label=metric, color=colors[i], edgecolor='white')
        ax.set_xticks(x + width)
        ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')
        ax.set_ylim(0, 1.05)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        if "ROSMAP" in title: ax.legend(fontsize=8, loc='upper right')

    plot_bar(ax_a1, df_brca, "BRCA Performance")
    plot_bar(ax_a2, df_rosmap, "ROSMAP Performance")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Heatmaps ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.3)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])
    
    sns.heatmap(df_brca.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Reds', fmt='.2f', cbar=False, ax=ax_b1, vmin=0, vmax=1)
    ax_b1.set_title("BRCA Heatmap", fontsize=10, fontweight='bold')
    sns.heatmap(df_rosmap.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Blues', fmt='.2f', cbar=False, ax=ax_b2, vmin=0, vmax=1)
    ax_b2.set_title("ROSMAP Heatmap", fontsize=10, fontweight='bold')
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Radar Charts ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.3)
    ax_c1 = fig.add_subplot(gs_c[0], polar=True)
    ax_c2 = fig.add_subplot(gs_c[1], polar=True)

    def plot_radar(ax, df, title):
        categories = ['Precision', 'Recall', 'F1-score']
        N = len(categories)
        angles = [n / float(N) * 2 * pi for n in range(N)]
        angles += angles[:1]
        ax.set_theta_offset(pi / 2); ax.set_theta_direction(-1)
        ax.set_xticks(angles[:-1]); ax.set_xticklabels(categories, fontsize=8)
        ax.set_ylim(0, 1); ax.set_yticks([0.2, 0.4, 0.6, 0.8]); ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=6)
        for index, row in df.iterrows():
            values = row[categories].values.flatten().tolist()
            values += values[:1]
            ax.plot(angles, values, linewidth=1.5, linestyle='solid', label=f"Cls {row['Class']}")
            ax.fill(angles, values, alpha=0.05)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=15)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fontsize=8, ncol=2)

    plot_radar(ax_c1, df_brca, "BRCA Profile")
    plot_radar(ax_c2, df_rosmap, "ROSMAP Profile")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Bubble Chart ---
    ax_d = fig.add_subplot(gs[1, 1])
    ax_d.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*10, alpha=0.6, c='#E74C3C', label='BRCA', edgecolors='black')
    ax_d.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*10, alpha=0.6, c='#3498DB', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_d.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    for i, txt in enumerate(df_rosmap['Class']):
        ax_d.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    ax_d.set_xlabel('Recall'); ax_d.set_ylabel('Precision'); ax_d.set_title("Precision vs Recall (Size = Support)", fontsize=10, fontweight='bold')
    ax_d.grid(True, linestyle='--', alpha=0.5); ax_d.legend(loc='upper left', fontsize=9)
    ax_d.text(-0.1, 1.05, "D", transform=ax_d.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # Title and Save
    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3) # Adjust top to make room for suptitle
    plt.savefig('figure_1_view.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 2. FIGURE 2: ANALYTICAL (Parallel, Lollipop, Marimekko, Slope)
# ==========================================
def generate_figure_2():
    print("Generating Figure 2: Analytical...")
    fig = plt.figure(figsize=(16, 10))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
    colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
    colors_rosmap = ['#E67E22', '#1ABC9C']

    # --- Panel A: Parallel Coords ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    
    def plot_pc(ax, df, cols, title):
        metrics = ['Precision', 'Recall', 'F1-score']
        x = range(len(metrics))
        for i, row in df.iterrows():
            ax.plot(x, row[metrics].values, marker='o', linewidth=2, label=f"Cls {row['Class']}", color=cols[i%len(cols)], alpha=0.8)
        ax.set_xticks(x); ax.set_xticklabels(metrics); ax.set_ylim(-0.05, 1.05)
        ax.set_title(title, fontweight='bold', fontsize=10); ax.grid(axis='y', linestyle='--', alpha=0.5)
        if "ROSMAP" in title: ax.set_yticklabels([]); ax.legend(loc='upper right', fontsize=8)
        else: ax.set_ylabel("Score"); ax.legend(loc='upper left', fontsize=8)

    plot_pc(ax_a1, df_brca, colors_brca, "BRCA Profiles")
    plot_pc(ax_a2, df_rosmap, colors_rosmap, "ROSMAP Profiles")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Lollipop Chart ---
    ax_b = fig.add_subplot(gs[0, 1])
    df_temp = pd.concat([df_brca.assign(D='BRCA'), df_rosmap.assign(D='ROSMAP')]).sort_values('F1-score')
    df_temp['Label'] = df_temp['D'] + " Cls " + df_temp['Class']
    y = range(len(df_temp))
    ax_b.hlines(y=y, xmin=0, xmax=df_temp['F1-score'], color='gray', alpha=0.5)
    for i, (val, label) in enumerate(zip(df_temp['F1-score'], df_temp['Label'])):
        col = '#E74C3C' if 'BRCA' in label else '#3498DB'
        ax_b.scatter(val, i, color=col, s=100, zorder=4)
        ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')
    ax_b.set_yticks(y); ax_b.set_yticklabels(df_temp['Label'])
    ax_b.set_title("Ranked Class Performance", fontweight='bold', fontsize=10); ax_b.set_xlabel("F1-Score")
    ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Marimekko ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_mari(ax, df, cols, title):
        curr_x = 0
        for i, row in df.iterrows():
            w, h = row['Support_Pct'], row['F1-score']
            rect = patches.Rectangle((curr_x, 0), w, h, linewidth=1, edgecolor='white', facecolor=cols[i%len(cols)], alpha=0.8)
            ax.add_patch(rect)
            if w > 0.1: ax.text(curr_x + w/2, h/2, f"C{row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
            curr_x += w
        ax.set_xlim(0, 1); ax.set_ylim(0, 1.05); ax.set_title(title, fontweight='bold', fontsize=10); ax.set_xlabel("Cumulative Support")
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("F1-Score")

    plot_mari(ax_c1, df_brca, colors_brca, "BRCA Impact Map")
    plot_mari(ax_c2, df_rosmap, colors_rosmap, "ROSMAP Impact Map")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Slope Graph ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])
    
    def plot_slope(ax, df, cols, title):
        for i, row in df.iterrows():
            ax.plot([0, 1], [row['Precision'], row['Recall']], color=cols[i%len(cols)], alpha=0.8, linewidth=2, marker='o')
        ax.set_xticks([0, 1]); ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
        ax.set_ylim(-0.05, 1.05); ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        ax.spines['left'].set_visible(False); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("Score")

    plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
    plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
    ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # Title and Save
    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_2_view.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 3. FIGURE 3: COMPARATIVE (Butterfly, Donut, Dumbbell, Dual-Axis)
# ==========================================
def generate_figure_3():
    print("Generating Figure 3: Comparative...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Butterfly ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_butterfly(ax, df, title):
        y = np.arange(len(df['Class']))
        ax.barh(y, -df['Precision'], color='#5DADE2', label='Precision', height=0.6)
        ax.barh(y, df['Recall'], color='#58D68D', label='Recall', height=0.6)
        ax.set_yticks(y); ax.set_yticklabels([f"Cls {c}" for c in df['Class']])
        ax.axvline(0, color='black', linewidth=0.8)
        ticks = np.linspace(-1, 1, 5)
        ax.set_xticks(ticks); ax.set_xticklabels([f"{abs(t):.1f}" for t in ticks], fontsize=8)
        ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=9)

    plot_butterfly(ax_a1, df_brca, "BRCA Metric Divergence")
    plot_butterfly(ax_a2, df_rosmap, "ROSMAP Metric Divergence")
    ax_a1.text(-0.2, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Donut ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.1)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_donut(ax, df, title):
        norm = mcolors.Normalize(vmin=0, vmax=0.8)
        cmap = cm.get_cmap('RdYlGn')
        colors = [cmap(norm(val)) for val in df['F1-score']]
        wedges, _ = ax.pie(df['Support'], colors=colors, startangle=90, wedgeprops=dict(width=0.4, edgecolor='white'))
        avg = np.average(df['F1-score'], weights=df['Support'])
        ax.text(0, 0, f"Avg F1\n{avg:.2f}", ha='center', va='center', fontweight='bold', fontsize=10)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
        ax.legend(wedges, [f"C{c}" for c in df['Class']], title="Classes", loc="center left", bbox_to_anchor=(0.85, 0, 0.5, 1), fontsize=8)

    plot_donut(ax_b1, df_brca, "BRCA (Color=F1)")
    plot_donut(ax_b2, df_rosmap, "ROSMAP (Color=F1)")
    ax_b1.text(-0.1, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Dumbbell ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.25)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_dumb(ax, df, title):
        for i, row in df.iterrows():
            ax.plot([row['Precision'], row['Recall']], [i, i], color='gray', zorder=1)
            ax.scatter(row['Precision'], i, color='#5DADE2', zorder=2, s=60)
            ax.scatter(row['Recall'], i, color='#58D68D', zorder=2, s=60)
        ax.set_yticks(range(len(df))); ax.set_yticklabels(df['Class'])
        ax.set_xlim(-0.1, 1.1); ax.set_title(title, fontsize=10, fontweight='bold'); ax.grid(axis='x', linestyle='--', alpha=0.5)

    plot_dumb(ax_c1, df_brca, "BRCA Gap Analysis")
    plot_dumb(ax_c2, df_rosmap, "ROSMAP Gap Analysis")
    ax_c1.text(-0.2, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Dual Axis ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.3)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_dual(ax, df, title):
        x = np.arange(len(df['Class']))
        ax.bar(x, df['Support'], color='lightgray', alpha=0.8, width=0.6)
        ax.set_ylabel('Support', color='gray', fontsize=8)
        ax2 = ax.twinx()
        ax2.plot(x, df['F1-score'], color='#E74C3C', marker='o', linewidth=2)
        ax2.set_ylabel('F1-Score', color='#E74C3C', fontsize=8)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dual(ax_d1, df_brca, "BRCA: Vol vs Qual")
    plot_dual(ax_d2, df_rosmap, "ROSMAP: Vol vs Qual")
    ax_d1.text(-0.25, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # Title and Save
    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_3_view.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 4. FIGURE 4: DIAGNOSTIC (Rose, Deviation, Zones, Outcome)
# ==========================================
def generate_figure_4():
    print("Generating Figure 4: Diagnostic...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Rose ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0], polar=True)
    ax_a2 = fig.add_subplot(gs_a[1], polar=True)
    
    def plot_rose(ax, df, title):
        N = len(df); theta = np.linspace(0.0, 2 * np.pi, N, endpoint=False); width = (2*np.pi) / N
        radii = df['F1-score']; colors = plt.cm.viridis(radii / 1.0)
        ax.bar(theta, radii, width=width, bottom=0.0, color=colors, alpha=0.8, edgecolor='white')
        ax.set_xticks(theta); ax.set_xticklabels([f"C{c}" for c in df['Class']], fontsize=9, fontweight='bold')
        ax.set_yticklabels([]); ax.set_ylim(0, 1); ax.set_title(title, fontsize=10, fontweight='bold', pad=20)

    plot_rose(ax_a1, df_brca, "BRCA: Performance Profile")
    plot_rose(ax_a2, df_rosmap, "ROSMAP: Performance Profile")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Deviation ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.25)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_dev(ax, df, avg, title):
        x = np.arange(len(df)); dev = df['F1-score'] - avg
        cols = ['#E74C3C' if d < 0 else '#2ECC71' for d in dev]
        ax.bar(x, dev, color=cols, alpha=0.8); ax.axhline(0, color='black', linewidth=1)
        for i, v in enumerate(dev):
            ax.text(i, v + (0.02 if v >=0 else -0.05), f"{v:+.2f}", ha='center', fontsize=8, fontweight='bold')
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_ylim(-0.8, 0.8); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dev(ax_b1, df_brca, brca_w_avg, "BRCA: Deviation")
    plot_dev(ax_b2, df_rosmap, rosmap_w_avg, "ROSMAP: Deviation")
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Zones ---
    ax_c = fig.add_subplot(gs[1, 0])
    ax_c.add_patch(patches.Rectangle((0, 0), 1, 1, color='#FDEDEC'))
    ax_c.add_patch(patches.Rectangle((0.6, 0.6), 0.4, 0.4, color='#EAFAF1'))
    ax_c.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*8, c='#3498DB', label='BRCA', edgecolors='black')
    ax_c.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*8, c='#E67E22', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_c.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, -10), textcoords='offset points', ha='center', fontsize=8)
    for i, txt in enumerate(df_rosmap['Class']):
        ax_c.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', ha='center', fontsize=8)
    ax_c.set_xlim(0, 1.05); ax_c.set_ylim(0, 1.05); ax_c.set_xlabel("Recall"); ax_c.set_ylabel("Precision")
    ax_c.set_title("Performance Assessment Zones", fontweight='bold', fontsize=10); ax_c.legend(loc='lower left')
    ax_c.text(-0.1, 1.1, "C", transform=ax_c.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Outcomes ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.25)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_out(ax, df, title):
        x = np.arange(len(df)); corr = df['Recall'] * df['Support']; err = df['Support'] - corr
        ax.bar(x, corr, color='#2ECC71', label='Correct', width=0.6)
        ax.bar(x, err, bottom=corr, color='#E74C3C', label='Error', width=0.6)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper right', fontsize=8)

    plot_out(ax_d1, df_brca, "BRCA Outcomes")
    plot_out(ax_d2, df_rosmap, "ROSMAP Outcomes")
    ax_d1.text(-0.2, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # Title and Save
    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_4_view.png', dpi=300, bbox_inches='tight')
    plt.close()

# ==========================================
# 5. EXECUTE ALL
# ==========================================
if __name__ == "__main__":
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    generate_figure_4()
    print("All 4 figures generated successfully with title: " + MAIN_TITLE)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from math import pi
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ==========================================
# 0. DATA PREPARATION (UPDATED FOR END-TO-END MODEL)
# ==========================================
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.55, 0.20, 0.80, 0.86, 0.83],
    'Recall':    [0.86, 0.08, 0.72, 0.88, 0.75],
    'F1-score':  [0.67, 0.12, 0.72, 0.87, 0.78],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.74, 0.73],
    'Recall':    [0.69, 0.78],
    'F1-score':  [0.71, 0.75],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Calculate percentages for Marimekko (Fig 2)
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# Calculate weighted averages for Deviation Plot (Fig 4)
brca_w_avg = np.average(df_brca['F1-score'], weights=df_brca['Support'])
rosmap_w_avg = np.average(df_rosmap['F1-score'], weights=df_rosmap['Support'])

# Global Plot Settings
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})
MAIN_TITLE = "Performance Evaluation of End-to-End Multi-Input DNN"


# ==========================================
# 1. FIGURE 1: OVERVIEW (Bar, Heatmap, Radar, Bubble)
# ==========================================
def generate_figure_1():
    print("Generating Figure 1: Overview...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Bar Charts ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.3)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    colors = ['#5DADE2', '#58D68D', '#EC7063']

    def plot_bar(ax, df, title):
        x = np.arange(len(df['Class']))
        width = 0.25
        metrics = ['Precision', 'Recall', 'F1-score']
        for i, metric in enumerate(metrics):
            ax.bar(x + i*width, df[metric], width=width, label=metric, color=colors[i], edgecolor='white')
        ax.set_xticks(x + width)
        ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')
        ax.set_ylim(0, 1.05)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        if "ROSMAP" in title: ax.legend(fontsize=8, loc='upper right')

    plot_bar(ax_a1, df_brca, "BRCA Performance")
    plot_bar(ax_a2, df_rosmap, "ROSMAP Performance")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Heatmaps ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.3)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])
    
    sns.heatmap(df_brca.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Reds', fmt='.2f', cbar=False, ax=ax_b1, vmin=0, vmax=1)
    ax_b1.set_title("BRCA Heatmap", fontsize=10, fontweight='bold')
    sns.heatmap(df_rosmap.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Blues', fmt='.2f', cbar=False, ax=ax_b2, vmin=0, vmax=1)
    ax_b2.set_title("ROSMAP Heatmap", fontsize=10, fontweight='bold')
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Radar Charts ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.3)
    ax_c1 = fig.add_subplot(gs_c[0], polar=True)
    ax_c2 = fig.add_subplot(gs_c[1], polar=True)

    def plot_radar(ax, df, title):
        categories = ['Precision', 'Recall', 'F1-score']
        N = len(categories)
        angles = [n / float(N) * 2 * pi for n in range(N)]
        angles += angles[:1]
        ax.set_theta_offset(pi / 2); ax.set_theta_direction(-1)
        ax.set_xticks(angles[:-1]); ax.set_xticklabels(categories, fontsize=8)
        ax.set_ylim(0, 1); ax.set_yticks([0.2, 0.4, 0.6, 0.8]); ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=6)
        for index, row in df.iterrows():
            values = row[categories].values.flatten().tolist()
            values += values[:1]
            ax.plot(angles, values, linewidth=1.5, linestyle='solid', label=f"Cls {row['Class']}")
            ax.fill(angles, values, alpha=0.05)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=15)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fontsize=8, ncol=2)

    plot_radar(ax_c1, df_brca, "BRCA Profile")
    plot_radar(ax_c2, df_rosmap, "ROSMAP Profile")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Bubble Chart ---
    ax_d = fig.add_subplot(gs[1, 1])
    ax_d.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*10, alpha=0.6, c='#E74C3C', label='BRCA', edgecolors='black')
    ax_d.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*10, alpha=0.6, c='#3498DB', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_d.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    for i, txt in enumerate(df_rosmap['Class']):
        ax_d.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    ax_d.set_xlabel('Recall'); ax_d.set_ylabel('Precision'); ax_d.set_title("Precision vs Recall (Size = Support)", fontsize=10, fontweight='bold')
    ax_d.grid(True, linestyle='--', alpha=0.5); ax_d.legend(loc='upper left', fontsize=9)
    ax_d.text(-0.1, 1.05, "D", transform=ax_d.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_1_end.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 2. FIGURE 2: ANALYTICAL (Parallel, Lollipop, Marimekko, Slope)
# ==========================================
def generate_figure_2():
    print("Generating Figure 2: Analytical...")
    fig = plt.figure(figsize=(16, 10))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
    colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
    colors_rosmap = ['#E67E22', '#1ABC9C']

    # --- Panel A: Parallel Coords ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    
    def plot_pc(ax, df, cols, title):
        metrics = ['Precision', 'Recall', 'F1-score']
        x = range(len(metrics))
        for i, row in df.iterrows():
            ax.plot(x, row[metrics].values, marker='o', linewidth=2, label=f"Cls {row['Class']}", color=cols[i%len(cols)], alpha=0.8)
        ax.set_xticks(x); ax.set_xticklabels(metrics); ax.set_ylim(-0.05, 1.05)
        ax.set_title(title, fontweight='bold', fontsize=10); ax.grid(axis='y', linestyle='--', alpha=0.5)
        if "ROSMAP" in title: ax.set_yticklabels([]); ax.legend(loc='upper right', fontsize=8)
        else: ax.set_ylabel("Score"); ax.legend(loc='upper left', fontsize=8)

    plot_pc(ax_a1, df_brca, colors_brca, "BRCA Profiles")
    plot_pc(ax_a2, df_rosmap, colors_rosmap, "ROSMAP Profiles")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Lollipop Chart ---
    ax_b = fig.add_subplot(gs[0, 1])
    df_temp = pd.concat([df_brca.assign(D='BRCA'), df_rosmap.assign(D='ROSMAP')]).sort_values('F1-score')
    df_temp['Label'] = df_temp['D'] + " Cls " + df_temp['Class']
    y = range(len(df_temp))
    ax_b.hlines(y=y, xmin=0, xmax=df_temp['F1-score'], color='gray', alpha=0.5)
    for i, (val, label) in enumerate(zip(df_temp['F1-score'], df_temp['Label'])):
        col = '#E74C3C' if 'BRCA' in label else '#3498DB'
        ax_b.scatter(val, i, color=col, s=100, zorder=4)
        ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')
    ax_b.set_yticks(y); ax_b.set_yticklabels(df_temp['Label'])
    ax_b.set_title("Ranked Class Performance", fontweight='bold', fontsize=10); ax_b.set_xlabel("F1-Score")
    ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Marimekko ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_mari(ax, df, cols, title):
        curr_x = 0
        for i, row in df.iterrows():
            w, h = row['Support_Pct'], row['F1-score']
            rect = patches.Rectangle((curr_x, 0), w, h, linewidth=1, edgecolor='white', facecolor=cols[i%len(cols)], alpha=0.8)
            ax.add_patch(rect)
            if w > 0.1: ax.text(curr_x + w/2, h/2, f"C{row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
            curr_x += w
        ax.set_xlim(0, 1); ax.set_ylim(0, 1.05); ax.set_title(title, fontweight='bold', fontsize=10); ax.set_xlabel("Cumulative Support")
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("F1-Score")

    plot_mari(ax_c1, df_brca, colors_brca, "BRCA Impact Map")
    plot_mari(ax_c2, df_rosmap, colors_rosmap, "ROSMAP Impact Map")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Slope Graph ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])
    
    def plot_slope(ax, df, cols, title):
        for i, row in df.iterrows():
            ax.plot([0, 1], [row['Precision'], row['Recall']], color=cols[i%len(cols)], alpha=0.8, linewidth=2, marker='o')
        ax.set_xticks([0, 1]); ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
        ax.set_ylim(-0.05, 1.05); ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        ax.spines['left'].set_visible(False); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("Score")

    plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
    plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
    ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_2_end.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 3. FIGURE 3: COMPARATIVE (Butterfly, Donut, Dumbbell, Dual-Axis)
# ==========================================
def generate_figure_3():
    print("Generating Figure 3: Comparative...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Butterfly ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_butterfly(ax, df, title):
        y = np.arange(len(df['Class']))
        ax.barh(y, -df['Precision'], color='#5DADE2', label='Precision', height=0.6)
        ax.barh(y, df['Recall'], color='#58D68D', label='Recall', height=0.6)
        ax.set_yticks(y); ax.set_yticklabels([f"Cls {c}" for c in df['Class']])
        ax.axvline(0, color='black', linewidth=0.8)
        ticks = np.linspace(-1, 1, 5)
        ax.set_xticks(ticks); ax.set_xticklabels([f"{abs(t):.1f}" for t in ticks], fontsize=8)
        ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=9)

    plot_butterfly(ax_a1, df_brca, "BRCA Metric Divergence")
    plot_butterfly(ax_a2, df_rosmap, "ROSMAP Metric Divergence")
    ax_a1.text(-0.2, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Donut ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.1)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_donut(ax, df, title):
        norm = mcolors.Normalize(vmin=0, vmax=0.8)
        cmap = cm.get_cmap('RdYlGn')
        colors = [cmap(norm(val)) for val in df['F1-score']]
        wedges, _ = ax.pie(df['Support'], colors=colors, startangle=90, wedgeprops=dict(width=0.4, edgecolor='white'))
        avg = np.average(df['F1-score'], weights=df['Support'])
        ax.text(0, 0, f"Avg F1\n{avg:.2f}", ha='center', va='center', fontweight='bold', fontsize=10)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
        ax.legend(wedges, [f"C{c}" for c in df['Class']], title="Classes", loc="center left", bbox_to_anchor=(0.85, 0, 0.5, 1), fontsize=8)

    plot_donut(ax_b1, df_brca, "BRCA (Color=F1)")
    plot_donut(ax_b2, df_rosmap, "ROSMAP (Color=F1)")
    ax_b1.text(-0.1, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Dumbbell ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.25)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_dumb(ax, df, title):
        for i, row in df.iterrows():
            ax.plot([row['Precision'], row['Recall']], [i, i], color='gray', zorder=1)
            ax.scatter(row['Precision'], i, color='#5DADE2', zorder=2, s=60)
            ax.scatter(row['Recall'], i, color='#58D68D', zorder=2, s=60)
        ax.set_yticks(range(len(df))); ax.set_yticklabels(df['Class'])
        ax.set_xlim(-0.1, 1.1); ax.set_title(title, fontsize=10, fontweight='bold'); ax.grid(axis='x', linestyle='--', alpha=0.5)

    plot_dumb(ax_c1, df_brca, "BRCA Gap Analysis")
    plot_dumb(ax_c2, df_rosmap, "ROSMAP Gap Analysis")
    ax_c1.text(-0.2, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Dual Axis ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.3)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_dual(ax, df, title):
        x = np.arange(len(df['Class']))
        ax.bar(x, df['Support'], color='lightgray', alpha=0.8, width=0.6)
        ax.set_ylabel('Support', color='gray', fontsize=8)
        ax2 = ax.twinx()
        ax2.plot(x, df['F1-score'], color='#E74C3C', marker='o', linewidth=2)
        ax2.set_ylabel('F1-Score', color='#E74C3C', fontsize=8)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dual(ax_d1, df_brca, "BRCA: Vol vs Qual")
    plot_dual(ax_d2, df_rosmap, "ROSMAP: Vol vs Qual")
    ax_d1.text(-0.25, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_3_end.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 4. FIGURE 4: DIAGNOSTIC (Rose, Deviation, Zones, Outcome)
# ==========================================
def generate_figure_4():
    print("Generating Figure 4: Diagnostic...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Rose ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0], polar=True)
    ax_a2 = fig.add_subplot(gs_a[1], polar=True)
    
    def plot_rose(ax, df, title):
        N = len(df); theta = np.linspace(0.0, 2 * np.pi, N, endpoint=False); width = (2*np.pi) / N
        radii = df['F1-score']; colors = plt.cm.viridis(radii / 1.0)
        ax.bar(theta, radii, width=width, bottom=0.0, color=colors, alpha=0.8, edgecolor='white')
        ax.set_xticks(theta); ax.set_xticklabels([f"C{c}" for c in df['Class']], fontsize=9, fontweight='bold')
        ax.set_yticklabels([]); ax.set_ylim(0, 1); ax.set_title(title, fontsize=10, fontweight='bold', pad=20)

    plot_rose(ax_a1, df_brca, "BRCA: Performance Profile")
    plot_rose(ax_a2, df_rosmap, "ROSMAP: Performance Profile")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Deviation ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.25)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_dev(ax, df, avg, title):
        x = np.arange(len(df)); dev = df['F1-score'] - avg
        cols = ['#E74C3C' if d < 0 else '#2ECC71' for d in dev]
        ax.bar(x, dev, color=cols, alpha=0.8); ax.axhline(0, color='black', linewidth=1)
        for i, v in enumerate(dev):
            ax.text(i, v + (0.02 if v >=0 else -0.05), f"{v:+.2f}", ha='center', fontsize=8, fontweight='bold')
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_ylim(-0.8, 0.8); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dev(ax_b1, df_brca, brca_w_avg, "BRCA: Deviation")
    plot_dev(ax_b2, df_rosmap, rosmap_w_avg, "ROSMAP: Deviation")
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Zones ---
    ax_c = fig.add_subplot(gs[1, 0])
    ax_c.add_patch(patches.Rectangle((0, 0), 1, 1, color='#FDEDEC'))
    ax_c.add_patch(patches.Rectangle((0.6, 0.6), 0.4, 0.4, color='#EAFAF1'))
    ax_c.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*8, c='#3498DB', label='BRCA', edgecolors='black')
    ax_c.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*8, c='#E67E22', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_c.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, -10), textcoords='offset points', ha='center', fontsize=8)
    for i, txt in enumerate(df_rosmap['Class']):
        ax_c.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', ha='center', fontsize=8)
    ax_c.set_xlim(0, 1.05); ax_c.set_ylim(0, 1.05); ax_c.set_xlabel("Recall"); ax_c.set_ylabel("Precision")
    ax_c.set_title("Performance Assessment Zones", fontweight='bold', fontsize=10); ax_c.legend(loc='lower left')
    ax_c.text(-0.1, 1.1, "C", transform=ax_c.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Outcomes ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.25)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_out(ax, df, title):
        x = np.arange(len(df)); corr = df['Recall'] * df['Support']; err = df['Support'] - corr
        ax.bar(x, correct, color='#2ECC71', label='Correct', width=0.6)
        # Note: 'correct' variable must be calculated inside function or passed
        # Fixing scope issue for this specific function block:
        corr = df['Recall'] * df['Support']
        err = df['Support'] - corr
        ax.bar(x, corr, color='#2ECC71', label='Correct', width=0.6)
        ax.bar(x, err, bottom=corr, color='#E74C3C', label='Error', width=0.6)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper right', fontsize=8)

    plot_out(ax_d1, df_brca, "BRCA Outcomes")
    plot_out(ax_d2, df_rosmap, "ROSMAP Outcomes")
    ax_d1.text(-0.2, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_4_end.png', dpi=300, bbox_inches='tight')
    plt.close()

# ==========================================
# 5. EXECUTE ALL
# ==========================================
if __name__ == "__main__":
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    generate_figure_4()
    print("All 4 figures generated successfully with title: " + MAIN_TITLE)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from math import pi
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ==========================================
# 0. DATA PREPARATION
# ==========================================
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.55, 0.20, 0.80, 0.86, 0.83],
    'Recall':    [0.86, 0.08, 0.72, 0.88, 0.75],
    'F1-score':  [0.67, 0.12, 0.72, 0.87, 0.78],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.74, 0.73],
    'Recall':    [0.69, 0.78],
    'F1-score':  [0.71, 0.75],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Calculate percentages for Marimekko (Fig 2)
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# Calculate weighted averages for Deviation Plot (Fig 4)
brca_w_avg = np.average(df_brca['F1-score'], weights=df_brca['Support'])
rosmap_w_avg = np.average(df_rosmap['F1-score'], weights=df_rosmap['Support'])

# Global Plot Settings
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})
MAIN_TITLE = "End-to-End Multi-Input Deep Neural Network (Late Fusion)"


# ==========================================
# 1. FIGURE 1: OVERVIEW
# ==========================================
def generate_figure_1():
    print("Generating Figure 1: Overview...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Bar Charts ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.3)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    colors = ['#5DADE2', '#58D68D', '#EC7063']

    def plot_bar(ax, df, title):
        x = np.arange(len(df['Class']))
        width = 0.25
        metrics = ['Precision', 'Recall', 'F1-score']
        for i, metric in enumerate(metrics):
            ax.bar(x + i*width, df[metric], width=width, label=metric, color=colors[i], edgecolor='white')
        ax.set_xticks(x + width)
        ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')
        ax.set_ylim(0, 1.05)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        if "ROSMAP" in title: ax.legend(fontsize=8, loc='upper right')

    plot_bar(ax_a1, df_brca, "BRCA Performance")
    plot_bar(ax_a2, df_rosmap, "ROSMAP Performance")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Heatmaps ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.3)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])
    
    sns.heatmap(df_brca.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Reds', fmt='.2f', cbar=False, ax=ax_b1, vmin=0, vmax=1)
    ax_b1.set_title("BRCA Heatmap", fontsize=10, fontweight='bold')
    sns.heatmap(df_rosmap.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Blues', fmt='.2f', cbar=False, ax=ax_b2, vmin=0, vmax=1)
    ax_b2.set_title("ROSMAP Heatmap", fontsize=10, fontweight='bold')
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Radar Charts ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.3)
    ax_c1 = fig.add_subplot(gs_c[0], polar=True)
    ax_c2 = fig.add_subplot(gs_c[1], polar=True)

    def plot_radar(ax, df, title):
        categories = ['Precision', 'Recall', 'F1-score']
        N = len(categories)
        angles = [n / float(N) * 2 * pi for n in range(N)]
        angles += angles[:1]
        ax.set_theta_offset(pi / 2); ax.set_theta_direction(-1)
        ax.set_xticks(angles[:-1]); ax.set_xticklabels(categories, fontsize=8)
        ax.set_ylim(0, 1); ax.set_yticks([0.2, 0.4, 0.6, 0.8]); ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=6)
        for index, row in df.iterrows():
            values = row[categories].values.flatten().tolist()
            values += values[:1]
            ax.plot(angles, values, linewidth=1.5, linestyle='solid', label=f"Cls {row['Class']}")
            ax.fill(angles, values, alpha=0.05)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=15)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fontsize=8, ncol=2)

    plot_radar(ax_c1, df_brca, "BRCA Profile")
    plot_radar(ax_c2, df_rosmap, "ROSMAP Profile")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Bubble Chart ---
    ax_d = fig.add_subplot(gs[1, 1])
    ax_d.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*10, alpha=0.6, c='#E74C3C', label='BRCA', edgecolors='black')
    ax_d.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*10, alpha=0.6, c='#3498DB', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_d.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    for i, txt in enumerate(df_rosmap['Class']):
        ax_d.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    ax_d.set_xlabel('Recall'); ax_d.set_ylabel('Precision'); ax_d.set_title("Precision vs Recall (Size = Support)", fontsize=10, fontweight='bold')
    ax_d.grid(True, linestyle='--', alpha=0.5); ax_d.legend(loc='upper left', fontsize=9)
    ax_d.text(-0.1, 1.05, "D", transform=ax_d.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_1_end.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 2. FIGURE 2: ANALYTICAL
# ==========================================
def generate_figure_2():
    print("Generating Figure 2: Analytical...")
    fig = plt.figure(figsize=(16, 10))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
    colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
    colors_rosmap = ['#E67E22', '#1ABC9C']

    # --- Panel A: Parallel Coords ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    
    def plot_pc(ax, df, cols, title):
        metrics = ['Precision', 'Recall', 'F1-score']
        x = range(len(metrics))
        for i, row in df.iterrows():
            ax.plot(x, row[metrics].values, marker='o', linewidth=2, label=f"Cls {row['Class']}", color=cols[i%len(cols)], alpha=0.8)
        ax.set_xticks(x); ax.set_xticklabels(metrics); ax.set_ylim(-0.05, 1.05)
        ax.set_title(title, fontweight='bold', fontsize=10); ax.grid(axis='y', linestyle='--', alpha=0.5)
        if "ROSMAP" in title: ax.set_yticklabels([]); ax.legend(loc='upper right', fontsize=8)
        else: ax.set_ylabel("Score"); ax.legend(loc='upper left', fontsize=8)

    plot_pc(ax_a1, df_brca, colors_brca, "BRCA Profiles")
    plot_pc(ax_a2, df_rosmap, colors_rosmap, "ROSMAP Profiles")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Lollipop Chart ---
    ax_b = fig.add_subplot(gs[0, 1])
    df_temp = pd.concat([df_brca.assign(D='BRCA'), df_rosmap.assign(D='ROSMAP')]).sort_values('F1-score')
    df_temp['Label'] = df_temp['D'] + " Cls " + df_temp['Class']
    y = range(len(df_temp))
    ax_b.hlines(y=y, xmin=0, xmax=df_temp['F1-score'], color='gray', alpha=0.5)
    for i, (val, label) in enumerate(zip(df_temp['F1-score'], df_temp['Label'])):
        col = '#E74C3C' if 'BRCA' in label else '#3498DB'
        ax_b.scatter(val, i, color=col, s=100, zorder=4)
        ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')
    ax_b.set_yticks(y); ax_b.set_yticklabels(df_temp['Label'])
    ax_b.set_title("Ranked Class Performance", fontweight='bold', fontsize=10); ax_b.set_xlabel("F1-Score")
    ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Marimekko ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_mari(ax, df, cols, title):
        curr_x = 0
        for i, row in df.iterrows():
            w, h = row['Support_Pct'], row['F1-score']
            rect = patches.Rectangle((curr_x, 0), w, h, linewidth=1, edgecolor='white', facecolor=cols[i%len(cols)], alpha=0.8)
            ax.add_patch(rect)
            if w > 0.1: ax.text(curr_x + w/2, h/2, f"C{row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
            curr_x += w
        ax.set_xlim(0, 1); ax.set_ylim(0, 1.05); ax.set_title(title, fontweight='bold', fontsize=10); ax.set_xlabel("Cumulative Support")
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("F1-Score")

    plot_mari(ax_c1, df_brca, colors_brca, "BRCA Impact Map")
    plot_mari(ax_c2, df_rosmap, colors_rosmap, "ROSMAP Impact Map")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Slope Graph ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])
    
    def plot_slope(ax, df, cols, title):
        for i, row in df.iterrows():
            ax.plot([0, 1], [row['Precision'], row['Recall']], color=cols[i%len(cols)], alpha=0.8, linewidth=2, marker='o')
        ax.set_xticks([0, 1]); ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
        ax.set_ylim(-0.05, 1.05); ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        ax.spines['left'].set_visible(False); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("Score")

    plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
    plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
    ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_2_end.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 3. FIGURE 3: COMPARATIVE
# ==========================================
def generate_figure_3():
    print("Generating Figure 3: Comparative...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Butterfly ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_butterfly(ax, df, title):
        y = np.arange(len(df['Class']))
        ax.barh(y, -df['Precision'], color='#5DADE2', label='Precision', height=0.6)
        ax.barh(y, df['Recall'], color='#58D68D', label='Recall', height=0.6)
        ax.set_yticks(y); ax.set_yticklabels([f"Cls {c}" for c in df['Class']])
        ax.axvline(0, color='black', linewidth=0.8)
        ticks = np.linspace(-1, 1, 5)
        ax.set_xticks(ticks); ax.set_xticklabels([f"{abs(t):.1f}" for t in ticks], fontsize=8)
        ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=9)

    plot_butterfly(ax_a1, df_brca, "BRCA Metric Divergence")
    plot_butterfly(ax_a2, df_rosmap, "ROSMAP Metric Divergence")
    ax_a1.text(-0.2, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Donut ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.1)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_donut(ax, df, title):
        norm = mcolors.Normalize(vmin=0, vmax=0.8)
        cmap = cm.get_cmap('RdYlGn')
        colors = [cmap(norm(val)) for val in df['F1-score']]
        wedges, _ = ax.pie(df['Support'], colors=colors, startangle=90, wedgeprops=dict(width=0.4, edgecolor='white'))
        avg = np.average(df['F1-score'], weights=df['Support'])
        ax.text(0, 0, f"Avg F1\n{avg:.2f}", ha='center', va='center', fontweight='bold', fontsize=10)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
        ax.legend(wedges, [f"C{c}" for c in df['Class']], title="Classes", loc="center left", bbox_to_anchor=(0.85, 0, 0.5, 1), fontsize=8)

    plot_donut(ax_b1, df_brca, "BRCA (Color=F1)")
    plot_donut(ax_b2, df_rosmap, "ROSMAP (Color=F1)")
    ax_b1.text(-0.1, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Dumbbell ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.25)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_dumb(ax, df, title):
        for i, row in df.iterrows():
            ax.plot([row['Precision'], row['Recall']], [i, i], color='gray', zorder=1)
            ax.scatter(row['Precision'], i, color='#5DADE2', zorder=2, s=60)
            ax.scatter(row['Recall'], i, color='#58D68D', zorder=2, s=60)
        ax.set_yticks(range(len(df))); ax.set_yticklabels(df['Class'])
        ax.set_xlim(-0.1, 1.1); ax.set_title(title, fontsize=10, fontweight='bold'); ax.grid(axis='x', linestyle='--', alpha=0.5)

    plot_dumb(ax_c1, df_brca, "BRCA Gap Analysis")
    plot_dumb(ax_c2, df_rosmap, "ROSMAP Gap Analysis")
    ax_c1.text(-0.2, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Dual Axis ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.3)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_dual(ax, df, title):
        x = np.arange(len(df['Class']))
        ax.bar(x, df['Support'], color='lightgray', alpha=0.8, width=0.6)
        ax.set_ylabel('Support', color='gray', fontsize=8)
        ax2 = ax.twinx()
        ax2.plot(x, df['F1-score'], color='#E74C3C', marker='o', linewidth=2)
        ax2.set_ylabel('F1-Score', color='#E74C3C', fontsize=8)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dual(ax_d1, df_brca, "BRCA: Vol vs Qual")
    plot_dual(ax_d2, df_rosmap, "ROSMAP: Vol vs Qual")
    ax_d1.text(-0.25, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_3_end.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 4. FIGURE 4: DIAGNOSTIC
# ==========================================
def generate_figure_4():
    print("Generating Figure 4: Diagnostic...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Rose ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0], polar=True)
    ax_a2 = fig.add_subplot(gs_a[1], polar=True)
    
    def plot_rose(ax, df, title):
        N = len(df); theta = np.linspace(0.0, 2 * np.pi, N, endpoint=False); width = (2*np.pi) / N
        radii = df['F1-score']; colors = plt.cm.viridis(radii / 1.0)
        ax.bar(theta, radii, width=width, bottom=0.0, color=colors, alpha=0.8, edgecolor='white')
        ax.set_xticks(theta); ax.set_xticklabels([f"C{c}" for c in df['Class']], fontsize=9, fontweight='bold')
        ax.set_yticklabels([]); ax.set_ylim(0, 1); ax.set_title(title, fontsize=10, fontweight='bold', pad=20)

    plot_rose(ax_a1, df_brca, "BRCA: Performance Profile")
    plot_rose(ax_a2, df_rosmap, "ROSMAP: Performance Profile")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Deviation ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.25)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_dev(ax, df, avg, title):
        x = np.arange(len(df)); dev = df['F1-score'] - avg
        cols = ['#E74C3C' if d < 0 else '#2ECC71' for d in dev]
        ax.bar(x, dev, color=cols, alpha=0.8); ax.axhline(0, color='black', linewidth=1)
        for i, v in enumerate(dev):
            ax.text(i, v + (0.02 if v >=0 else -0.05), f"{v:+.2f}", ha='center', fontsize=8, fontweight='bold')
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_ylim(-0.8, 0.8); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dev(ax_b1, df_brca, brca_w_avg, "BRCA: Deviation")
    plot_dev(ax_b2, df_rosmap, rosmap_w_avg, "ROSMAP: Deviation")
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Zones ---
    ax_c = fig.add_subplot(gs[1, 0])
    ax_c.add_patch(patches.Rectangle((0, 0), 1, 1, color='#FDEDEC'))
    ax_c.add_patch(patches.Rectangle((0.6, 0.6), 0.4, 0.4, color='#EAFAF1'))
    ax_c.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*8, c='#3498DB', label='BRCA', edgecolors='black')
    ax_c.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*8, c='#E67E22', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_c.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, -10), textcoords='offset points', ha='center', fontsize=8)
    for i, txt in enumerate(df_rosmap['Class']):
        ax_c.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', ha='center', fontsize=8)
    ax_c.set_xlim(0, 1.05); ax_c.set_ylim(0, 1.05); ax_c.set_xlabel("Recall"); ax_c.set_ylabel("Precision")
    ax_c.set_title("Performance Assessment Zones", fontweight='bold', fontsize=10); ax_c.legend(loc='lower left')
    ax_c.text(-0.1, 1.1, "C", transform=ax_c.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Outcomes ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.25)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_out(ax, df, title):
        x = np.arange(len(df))
        # Correctly calculate values using df columns inside the function
        corr = df['Recall'] * df['Support']
        err = df['Support'] - corr
        
        # Plotting
        ax.bar(x, corr, color='#2ECC71', label='Correct', width=0.6)
        ax.bar(x, err, bottom=corr, color='#E74C3C', label='Error', width=0.6)
        
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper right', fontsize=8)

    plot_out(ax_d1, df_brca, "BRCA Outcomes")
    plot_out(ax_d2, df_rosmap, "ROSMAP Outcomes")
    ax_d1.text(-0.2, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_4_end.png', dpi=300, bbox_inches='tight')
    plt.close()

# ==========================================
# 5. EXECUTE ALL
# ==========================================
if __name__ == "__main__":
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    generate_figure_4()
    print("All 4 figures generated successfully with title: " + MAIN_TITLE)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from math import pi
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ==========================================
# 0. DATA PREPARATION (UPDATED FOR TWO-STAGE EARLY FUSION)
# ==========================================
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.57, 0.97, 0.85, 0.89, 0.85],
    'Recall':    [0.66, 0.90, 0.79, 0.90, 0.80],
    'F1-score':  [0.61, 0.93, 0.81, 0.89, 0.82],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.77, 0.86],
    'Recall':    [0.86, 0.76],
    'F1-score':  [0.81, 0.81],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Calculate percentages for Marimekko (Fig 2)
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# Calculate weighted averages for Deviation Plot (Fig 4)
brca_w_avg = np.average(df_brca['F1-score'], weights=df_brca['Support'])
rosmap_w_avg = np.average(df_rosmap['F1-score'], weights=df_rosmap['Support'])

# Global Plot Settings
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})
MAIN_TITLE = "Two-Stage Early Fusion Network"


# ==========================================
# 1. FIGURE 1: OVERVIEW
# ==========================================
def generate_figure_1():
    print("Generating Figure 1: Overview...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Bar Charts ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.3)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    colors = ['#5DADE2', '#58D68D', '#EC7063']

    def plot_bar(ax, df, title):
        x = np.arange(len(df['Class']))
        width = 0.25
        metrics = ['Precision', 'Recall', 'F1-score']
        for i, metric in enumerate(metrics):
            ax.bar(x + i*width, df[metric], width=width, label=metric, color=colors[i], edgecolor='white')
        ax.set_xticks(x + width)
        ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')
        ax.set_ylim(0, 1.05)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        if "ROSMAP" in title: ax.legend(fontsize=8, loc='upper right')

    plot_bar(ax_a1, df_brca, "BRCA Performance")
    plot_bar(ax_a2, df_rosmap, "ROSMAP Performance")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Heatmaps ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.3)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])
    
    sns.heatmap(df_brca.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Reds', fmt='.2f', cbar=False, ax=ax_b1, vmin=0, vmax=1)
    ax_b1.set_title("BRCA Heatmap", fontsize=10, fontweight='bold')
    sns.heatmap(df_rosmap.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Blues', fmt='.2f', cbar=False, ax=ax_b2, vmin=0, vmax=1)
    ax_b2.set_title("ROSMAP Heatmap", fontsize=10, fontweight='bold')
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Radar Charts ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.3)
    ax_c1 = fig.add_subplot(gs_c[0], polar=True)
    ax_c2 = fig.add_subplot(gs_c[1], polar=True)

    def plot_radar(ax, df, title):
        categories = ['Precision', 'Recall', 'F1-score']
        N = len(categories)
        angles = [n / float(N) * 2 * pi for n in range(N)]
        angles += angles[:1]
        ax.set_theta_offset(pi / 2); ax.set_theta_direction(-1)
        ax.set_xticks(angles[:-1]); ax.set_xticklabels(categories, fontsize=8)
        ax.set_ylim(0, 1); ax.set_yticks([0.2, 0.4, 0.6, 0.8]); ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=6)
        for index, row in df.iterrows():
            values = row[categories].values.flatten().tolist()
            values += values[:1]
            ax.plot(angles, values, linewidth=1.5, linestyle='solid', label=f"Cls {row['Class']}")
            ax.fill(angles, values, alpha=0.05)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=15)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fontsize=8, ncol=2)

    plot_radar(ax_c1, df_brca, "BRCA Profile")
    plot_radar(ax_c2, df_rosmap, "ROSMAP Profile")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Bubble Chart ---
    ax_d = fig.add_subplot(gs[1, 1])
    ax_d.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*10, alpha=0.6, c='#E74C3C', label='BRCA', edgecolors='black')
    ax_d.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*10, alpha=0.6, c='#3498DB', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_d.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    for i, txt in enumerate(df_rosmap['Class']):
        ax_d.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    ax_d.set_xlabel('Recall'); ax_d.set_ylabel('Precision'); ax_d.set_title("Precision vs Recall (Size = Support)", fontsize=10, fontweight='bold')
    ax_d.grid(True, linestyle='--', alpha=0.5); ax_d.legend(loc='upper left', fontsize=9)
    ax_d.text(-0.1, 1.05, "D", transform=ax_d.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_1_early_fusion.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 2. FIGURE 2: ANALYTICAL
# ==========================================
def generate_figure_2():
    print("Generating Figure 2: Analytical...")
    fig = plt.figure(figsize=(16, 10))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
    colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
    colors_rosmap = ['#E67E22', '#1ABC9C']

    # --- Panel A: Parallel Coords ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    
    def plot_pc(ax, df, cols, title):
        metrics = ['Precision', 'Recall', 'F1-score']
        x = range(len(metrics))
        for i, row in df.iterrows():
            ax.plot(x, row[metrics].values, marker='o', linewidth=2, label=f"Cls {row['Class']}", color=cols[i%len(cols)], alpha=0.8)
        ax.set_xticks(x); ax.set_xticklabels(metrics); ax.set_ylim(-0.05, 1.05)
        ax.set_title(title, fontweight='bold', fontsize=10); ax.grid(axis='y', linestyle='--', alpha=0.5)
        if "ROSMAP" in title: ax.set_yticklabels([]); ax.legend(loc='upper right', fontsize=8)
        else: ax.set_ylabel("Score"); ax.legend(loc='upper left', fontsize=8)

    plot_pc(ax_a1, df_brca, colors_brca, "BRCA Profiles")
    plot_pc(ax_a2, df_rosmap, colors_rosmap, "ROSMAP Profiles")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Lollipop Chart ---
    ax_b = fig.add_subplot(gs[0, 1])
    df_temp = pd.concat([df_brca.assign(D='BRCA'), df_rosmap.assign(D='ROSMAP')]).sort_values('F1-score')
    df_temp['Label'] = df_temp['D'] + " Cls " + df_temp['Class']
    y = range(len(df_temp))
    ax_b.hlines(y=y, xmin=0, xmax=df_temp['F1-score'], color='gray', alpha=0.5)
    for i, (val, label) in enumerate(zip(df_temp['F1-score'], df_temp['Label'])):
        col = '#E74C3C' if 'BRCA' in label else '#3498DB'
        ax_b.scatter(val, i, color=col, s=100, zorder=4)
        ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')
    ax_b.set_yticks(y); ax_b.set_yticklabels(df_temp['Label'])
    ax_b.set_title("Ranked Class Performance", fontweight='bold', fontsize=10); ax_b.set_xlabel("F1-Score")
    ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Marimekko ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_mari(ax, df, cols, title):
        curr_x = 0
        for i, row in df.iterrows():
            w, h = row['Support_Pct'], row['F1-score']
            rect = patches.Rectangle((curr_x, 0), w, h, linewidth=1, edgecolor='white', facecolor=cols[i%len(cols)], alpha=0.8)
            ax.add_patch(rect)
            if w > 0.1: ax.text(curr_x + w/2, h/2, f"C{row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
            curr_x += w
        ax.set_xlim(0, 1); ax.set_ylim(0, 1.05); ax.set_title(title, fontweight='bold', fontsize=10); ax.set_xlabel("Cumulative Support")
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("F1-Score")

    plot_mari(ax_c1, df_brca, colors_brca, "BRCA Impact Map")
    plot_mari(ax_c2, df_rosmap, colors_rosmap, "ROSMAP Impact Map")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Slope Graph ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])
    
    def plot_slope(ax, df, cols, title):
        for i, row in df.iterrows():
            ax.plot([0, 1], [row['Precision'], row['Recall']], color=cols[i%len(cols)], alpha=0.8, linewidth=2, marker='o')
        ax.set_xticks([0, 1]); ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
        ax.set_ylim(-0.05, 1.05); ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        ax.spines['left'].set_visible(False); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("Score")

    plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
    plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
    ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_2_early_fusion.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 3. FIGURE 3: COMPARATIVE
# ==========================================
def generate_figure_3():
    print("Generating Figure 3: Comparative...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Butterfly ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_butterfly(ax, df, title):
        y = np.arange(len(df['Class']))
        ax.barh(y, -df['Precision'], color='#5DADE2', label='Precision', height=0.6)
        ax.barh(y, df['Recall'], color='#58D68D', label='Recall', height=0.6)
        ax.set_yticks(y); ax.set_yticklabels([f"Cls {c}" for c in df['Class']])
        ax.axvline(0, color='black', linewidth=0.8)
        ticks = np.linspace(-1, 1, 5)
        ax.set_xticks(ticks); ax.set_xticklabels([f"{abs(t):.1f}" for t in ticks], fontsize=8)
        ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=9)

    plot_butterfly(ax_a1, df_brca, "BRCA Metric Divergence")
    plot_butterfly(ax_a2, df_rosmap, "ROSMAP Metric Divergence")
    ax_a1.text(-0.2, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Donut ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.1)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_donut(ax, df, title):
        norm = mcolors.Normalize(vmin=0, vmax=0.8)
        cmap = cm.get_cmap('RdYlGn')
        colors = [cmap(norm(val)) for val in df['F1-score']]
        wedges, _ = ax.pie(df['Support'], colors=colors, startangle=90, wedgeprops=dict(width=0.4, edgecolor='white'))
        avg = np.average(df['F1-score'], weights=df['Support'])
        ax.text(0, 0, f"Avg F1\n{avg:.2f}", ha='center', va='center', fontweight='bold', fontsize=10)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
        ax.legend(wedges, [f"C{c}" for c in df['Class']], title="Classes", loc="center left", bbox_to_anchor=(0.85, 0, 0.5, 1), fontsize=8)

    plot_donut(ax_b1, df_brca, "BRCA (Color=F1)")
    plot_donut(ax_b2, df_rosmap, "ROSMAP (Color=F1)")
    ax_b1.text(-0.1, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Dumbbell ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.25)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_dumb(ax, df, title):
        for i, row in df.iterrows():
            ax.plot([row['Precision'], row['Recall']], [i, i], color='gray', zorder=1)
            ax.scatter(row['Precision'], i, color='#5DADE2', zorder=2, s=60)
            ax.scatter(row['Recall'], i, color='#58D68D', zorder=2, s=60)
        ax.set_yticks(range(len(df))); ax.set_yticklabels(df['Class'])
        ax.set_xlim(-0.1, 1.1); ax.set_title(title, fontsize=10, fontweight='bold'); ax.grid(axis='x', linestyle='--', alpha=0.5)

    plot_dumb(ax_c1, df_brca, "BRCA Gap Analysis")
    plot_dumb(ax_c2, df_rosmap, "ROSMAP Gap Analysis")
    ax_c1.text(-0.2, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Dual Axis ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.3)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_dual(ax, df, title):
        x = np.arange(len(df['Class']))
        ax.bar(x, df['Support'], color='lightgray', alpha=0.8, width=0.6)
        ax.set_ylabel('Support', color='gray', fontsize=8)
        ax2 = ax.twinx()
        ax2.plot(x, df['F1-score'], color='#E74C3C', marker='o', linewidth=2)
        ax2.set_ylabel('F1-Score', color='#E74C3C', fontsize=8)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dual(ax_d1, df_brca, "BRCA: Vol vs Qual")
    plot_dual(ax_d2, df_rosmap, "ROSMAP: Vol vs Qual")
    ax_d1.text(-0.25, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_3_early_fusion.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 4. FIGURE 4: DIAGNOSTIC
# ==========================================
def generate_figure_4():
    print("Generating Figure 4: Diagnostic...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Rose ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0], polar=True)
    ax_a2 = fig.add_subplot(gs_a[1], polar=True)
    
    def plot_rose(ax, df, title):
        N = len(df); theta = np.linspace(0.0, 2 * np.pi, N, endpoint=False); width = (2*np.pi) / N
        radii = df['F1-score']; colors = plt.cm.viridis(radii / 1.0)
        ax.bar(theta, radii, width=width, bottom=0.0, color=colors, alpha=0.8, edgecolor='white')
        ax.set_xticks(theta); ax.set_xticklabels([f"C{c}" for c in df['Class']], fontsize=9, fontweight='bold')
        ax.set_yticklabels([]); ax.set_ylim(0, 1); ax.set_title(title, fontsize=10, fontweight='bold', pad=20)

    plot_rose(ax_a1, df_brca, "BRCA: Performance Profile")
    plot_rose(ax_a2, df_rosmap, "ROSMAP: Performance Profile")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Deviation ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.25)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_dev(ax, df, avg, title):
        x = np.arange(len(df)); dev = df['F1-score'] - avg
        cols = ['#E74C3C' if d < 0 else '#2ECC71' for d in dev]
        ax.bar(x, dev, color=cols, alpha=0.8); ax.axhline(0, color='black', linewidth=1)
        for i, v in enumerate(dev):
            ax.text(i, v + (0.02 if v >=0 else -0.05), f"{v:+.2f}", ha='center', fontsize=8, fontweight='bold')
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_ylim(-0.8, 0.8); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dev(ax_b1, df_brca, brca_w_avg, "BRCA: Deviation")
    plot_dev(ax_b2, df_rosmap, rosmap_w_avg, "ROSMAP: Deviation")
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Zones ---
    ax_c = fig.add_subplot(gs[1, 0])
    ax_c.add_patch(patches.Rectangle((0, 0), 1, 1, color='#FDEDEC'))
    ax_c.add_patch(patches.Rectangle((0.6, 0.6), 0.4, 0.4, color='#EAFAF1'))
    ax_c.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*8, c='#3498DB', label='BRCA', edgecolors='black')
    ax_c.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*8, c='#E67E22', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_c.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, -10), textcoords='offset points', ha='center', fontsize=8)
    for i, txt in enumerate(df_rosmap['Class']):
        ax_c.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', ha='center', fontsize=8)
    ax_c.set_xlim(0, 1.05); ax_c.set_ylim(0, 1.05); ax_c.set_xlabel("Recall"); ax_c.set_ylabel("Precision")
    ax_c.set_title("Performance Assessment Zones", fontweight='bold', fontsize=10); ax_c.legend(loc='lower left')
    ax_c.text(-0.1, 1.1, "C", transform=ax_c.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Outcomes ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.25)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_out(ax, df, title):
        x = np.arange(len(df))
        # Correctly calculate values using df columns inside the function
        corr = df['Recall'] * df['Support']
        err = df['Support'] - corr
        
        # Plotting
        ax.bar(x, corr, color='#2ECC71', label='Correct', width=0.6)
        ax.bar(x, err, bottom=corr, color='#E74C3C', label='Error', width=0.6)
        
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper right', fontsize=8)

    plot_out(ax_d1, df_brca, "BRCA Outcomes")
    plot_out(ax_d2, df_rosmap, "ROSMAP Outcomes")
    ax_d1.text(-0.2, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_4_early_fusion.png', dpi=300, bbox_inches='tight')
    plt.close()

# ==========================================
# 5. EXECUTE ALL
# ==========================================
if __name__ == "__main__":
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    generate_figure_4()
    print("All 4 figures generated successfully with title: " + MAIN_TITLE)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from math import pi
import matplotlib.gridspec as gridspec
import matplotlib.patches as patches
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ==========================================
# 0. DATA PREPARATION (UPDATED FOR HYBRID FUSION)
# ==========================================
brca_data = {
    'Class': ['0.0', '1.0', '2.0', '3.0', '4.0'],
    'Precision': [0.60, 0.97, 0.75, 0.96, 0.89],
    'Recall':    [0.80, 0.90, 0.80, 0.90, 0.89],
    'F1-score':  [0.68, 0.93, 0.77, 0.92, 0.89],
    'Support':   [35, 39, 14, 131, 44]
}

rosmap_data = {
    'Class': ['0.0', '1.0'],
    'Precision': [0.82, 0.82],
    'Recall':    [0.80, 0.84],
    'F1-score':  [0.81, 0.83],
    'Support':   [51, 55]
}

df_brca = pd.DataFrame(brca_data)
df_rosmap = pd.DataFrame(rosmap_data)

# Calculate percentages for Marimekko (Fig 2)
df_brca['Support_Pct'] = df_brca['Support'] / df_brca['Support'].sum()
df_rosmap['Support_Pct'] = df_rosmap['Support'] / df_rosmap['Support'].sum()

# Calculate weighted averages for Deviation Plot (Fig 4)
brca_w_avg = np.average(df_brca['F1-score'], weights=df_brca['Support'])
rosmap_w_avg = np.average(df_rosmap['F1-score'], weights=df_rosmap['Support'])

# Global Plot Settings
plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial'], 'font.size': 10})
MAIN_TITLE = "Hybrid Latent-Original Feature Fusion Network"


# ==========================================
# 1. FIGURE 1: OVERVIEW
# ==========================================
def generate_figure_1():
    print("Generating Figure 1: Overview...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Bar Charts ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.3)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    colors = ['#5DADE2', '#58D68D', '#EC7063']

    def plot_bar(ax, df, title):
        x = np.arange(len(df['Class']))
        width = 0.25
        metrics = ['Precision', 'Recall', 'F1-score']
        for i, metric in enumerate(metrics):
            ax.bar(x + i*width, df[metric], width=width, label=metric, color=colors[i], edgecolor='white')
        ax.set_xticks(x + width)
        ax.set_xticklabels(df['Class'])
        ax.set_title(title, fontsize=10, fontweight='bold')
        ax.set_ylim(0, 1.05)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        if "ROSMAP" in title: ax.legend(fontsize=8, loc='upper right')

    plot_bar(ax_a1, df_brca, "BRCA Performance")
    plot_bar(ax_a2, df_rosmap, "ROSMAP Performance")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Heatmaps ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.3)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])
    
    sns.heatmap(df_brca.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Reds', fmt='.2f', cbar=False, ax=ax_b1, vmin=0, vmax=1)
    ax_b1.set_title("BRCA Heatmap", fontsize=10, fontweight='bold')
    sns.heatmap(df_rosmap.set_index('Class')[['Precision', 'Recall', 'F1-score']], annot=True, cmap='Blues', fmt='.2f', cbar=False, ax=ax_b2, vmin=0, vmax=1)
    ax_b2.set_title("ROSMAP Heatmap", fontsize=10, fontweight='bold')
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Radar Charts ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.3)
    ax_c1 = fig.add_subplot(gs_c[0], polar=True)
    ax_c2 = fig.add_subplot(gs_c[1], polar=True)

    def plot_radar(ax, df, title):
        categories = ['Precision', 'Recall', 'F1-score']
        N = len(categories)
        angles = [n / float(N) * 2 * pi for n in range(N)]
        angles += angles[:1]
        ax.set_theta_offset(pi / 2); ax.set_theta_direction(-1)
        ax.set_xticks(angles[:-1]); ax.set_xticklabels(categories, fontsize=8)
        ax.set_ylim(0, 1); ax.set_yticks([0.2, 0.4, 0.6, 0.8]); ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=6)
        for index, row in df.iterrows():
            values = row[categories].values.flatten().tolist()
            values += values[:1]
            ax.plot(angles, values, linewidth=1.5, linestyle='solid', label=f"Cls {row['Class']}")
            ax.fill(angles, values, alpha=0.05)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=15)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fontsize=8, ncol=2)

    plot_radar(ax_c1, df_brca, "BRCA Profile")
    plot_radar(ax_c2, df_rosmap, "ROSMAP Profile")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Bubble Chart ---
    ax_d = fig.add_subplot(gs[1, 1])
    ax_d.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*10, alpha=0.6, c='#E74C3C', label='BRCA', edgecolors='black')
    ax_d.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*10, alpha=0.6, c='#3498DB', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_d.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    for i, txt in enumerate(df_rosmap['Class']):
        ax_d.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', fontsize=8, ha='center')
    ax_d.set_xlabel('Recall'); ax_d.set_ylabel('Precision'); ax_d.set_title("Precision vs Recall (Size = Support)", fontsize=10, fontweight='bold')
    ax_d.grid(True, linestyle='--', alpha=0.5); ax_d.legend(loc='upper left', fontsize=9)
    ax_d.text(-0.1, 1.05, "D", transform=ax_d.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_1_hybrid.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 2. FIGURE 2: ANALYTICAL
# ==========================================
def generate_figure_2():
    print("Generating Figure 2: Analytical...")
    fig = plt.figure(figsize=(16, 10))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])
    colors_brca = ['#E74C3C', '#9B59B6', '#3498DB', '#2ECC71', '#F1C40F']
    colors_rosmap = ['#E67E22', '#1ABC9C']

    # --- Panel A: Parallel Coords ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.15)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])
    
    def plot_pc(ax, df, cols, title):
        metrics = ['Precision', 'Recall', 'F1-score']
        x = range(len(metrics))
        for i, row in df.iterrows():
            ax.plot(x, row[metrics].values, marker='o', linewidth=2, label=f"Cls {row['Class']}", color=cols[i%len(cols)], alpha=0.8)
        ax.set_xticks(x); ax.set_xticklabels(metrics); ax.set_ylim(-0.05, 1.05)
        ax.set_title(title, fontweight='bold', fontsize=10); ax.grid(axis='y', linestyle='--', alpha=0.5)
        if "ROSMAP" in title: ax.set_yticklabels([]); ax.legend(loc='upper right', fontsize=8)
        else: ax.set_ylabel("Score"); ax.legend(loc='upper left', fontsize=8)

    plot_pc(ax_a1, df_brca, colors_brca, "BRCA Profiles")
    plot_pc(ax_a2, df_rosmap, colors_rosmap, "ROSMAP Profiles")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Lollipop Chart ---
    ax_b = fig.add_subplot(gs[0, 1])
    df_temp = pd.concat([df_brca.assign(D='BRCA'), df_rosmap.assign(D='ROSMAP')]).sort_values('F1-score')
    df_temp['Label'] = df_temp['D'] + " Cls " + df_temp['Class']
    y = range(len(df_temp))
    ax_b.hlines(y=y, xmin=0, xmax=df_temp['F1-score'], color='gray', alpha=0.5)
    for i, (val, label) in enumerate(zip(df_temp['F1-score'], df_temp['Label'])):
        col = '#E74C3C' if 'BRCA' in label else '#3498DB'
        ax_b.scatter(val, i, color=col, s=100, zorder=4)
        ax_b.text(val + 0.02, i, f"{val:.2f}", va='center', fontsize=9, fontweight='bold')
    ax_b.set_yticks(y); ax_b.set_yticklabels(df_temp['Label'])
    ax_b.set_title("Ranked Class Performance", fontweight='bold', fontsize=10); ax_b.set_xlabel("F1-Score")
    ax_b.text(-0.1, 1.1, "B", transform=ax_b.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Marimekko ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.15)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_mari(ax, df, cols, title):
        curr_x = 0
        for i, row in df.iterrows():
            w, h = row['Support_Pct'], row['F1-score']
            rect = patches.Rectangle((curr_x, 0), w, h, linewidth=1, edgecolor='white', facecolor=cols[i%len(cols)], alpha=0.8)
            ax.add_patch(rect)
            if w > 0.1: ax.text(curr_x + w/2, h/2, f"C{row['Class']}", ha='center', va='center', color='white', fontweight='bold', fontsize=8)
            curr_x += w
        ax.set_xlim(0, 1); ax.set_ylim(0, 1.05); ax.set_title(title, fontweight='bold', fontsize=10); ax.set_xlabel("Cumulative Support")
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("F1-Score")

    plot_mari(ax_c1, df_brca, colors_brca, "BRCA Impact Map")
    plot_mari(ax_c2, df_rosmap, colors_rosmap, "ROSMAP Impact Map")
    ax_c1.text(-0.15, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Slope Graph ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.15)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])
    
    def plot_slope(ax, df, cols, title):
        for i, row in df.iterrows():
            ax.plot([0, 1], [row['Precision'], row['Recall']], color=cols[i%len(cols)], alpha=0.8, linewidth=2, marker='o')
        ax.set_xticks([0, 1]); ax.set_xticklabels(['Precision', 'Recall'], fontweight='bold')
        ax.set_ylim(-0.05, 1.05); ax.set_title(title, fontweight='bold', fontsize=10)
        ax.grid(axis='y', linestyle=':', alpha=0.5)
        ax.spines['left'].set_visible(False); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
        if "ROSMAP" in title: ax.set_yticklabels([])
        else: ax.set_ylabel("Score")

    plot_slope(ax_d1, df_brca, colors_brca, "BRCA Trade-offs")
    plot_slope(ax_d2, df_rosmap, colors_rosmap, "ROSMAP Trade-offs")
    ax_d1.text(-0.15, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_2_hybrid.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 3. FIGURE 3: COMPARATIVE
# ==========================================
def generate_figure_3():
    print("Generating Figure 3: Comparative...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Butterfly ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0])
    ax_a2 = fig.add_subplot(gs_a[1])

    def plot_butterfly(ax, df, title):
        y = np.arange(len(df['Class']))
        ax.barh(y, -df['Precision'], color='#5DADE2', label='Precision', height=0.6)
        ax.barh(y, df['Recall'], color='#58D68D', label='Recall', height=0.6)
        ax.set_yticks(y); ax.set_yticklabels([f"Cls {c}" for c in df['Class']])
        ax.axvline(0, color='black', linewidth=0.8)
        ticks = np.linspace(-1, 1, 5)
        ax.set_xticks(ticks); ax.set_xticklabels([f"{abs(t):.1f}" for t in ticks], fontsize=8)
        ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=9)

    plot_butterfly(ax_a1, df_brca, "BRCA Metric Divergence")
    plot_butterfly(ax_a2, df_rosmap, "ROSMAP Metric Divergence")
    ax_a1.text(-0.2, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Donut ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.1)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_donut(ax, df, title):
        norm = mcolors.Normalize(vmin=0, vmax=0.8)
        cmap = cm.get_cmap('RdYlGn')
        colors = [cmap(norm(val)) for val in df['F1-score']]
        wedges, _ = ax.pie(df['Support'], colors=colors, startangle=90, wedgeprops=dict(width=0.4, edgecolor='white'))
        avg = np.average(df['F1-score'], weights=df['Support'])
        ax.text(0, 0, f"Avg F1\n{avg:.2f}", ha='center', va='center', fontweight='bold', fontsize=10)
        ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
        ax.legend(wedges, [f"C{c}" for c in df['Class']], title="Classes", loc="center left", bbox_to_anchor=(0.85, 0, 0.5, 1), fontsize=8)

    plot_donut(ax_b1, df_brca, "BRCA (Color=F1)")
    plot_donut(ax_b2, df_rosmap, "ROSMAP (Color=F1)")
    ax_b1.text(-0.1, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Dumbbell ---
    gs_c = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 0], wspace=0.25)
    ax_c1 = fig.add_subplot(gs_c[0])
    ax_c2 = fig.add_subplot(gs_c[1])
    
    def plot_dumb(ax, df, title):
        for i, row in df.iterrows():
            ax.plot([row['Precision'], row['Recall']], [i, i], color='gray', zorder=1)
            ax.scatter(row['Precision'], i, color='#5DADE2', zorder=2, s=60)
            ax.scatter(row['Recall'], i, color='#58D68D', zorder=2, s=60)
        ax.set_yticks(range(len(df))); ax.set_yticklabels(df['Class'])
        ax.set_xlim(-0.1, 1.1); ax.set_title(title, fontsize=10, fontweight='bold'); ax.grid(axis='x', linestyle='--', alpha=0.5)

    plot_dumb(ax_c1, df_brca, "BRCA Gap Analysis")
    plot_dumb(ax_c2, df_rosmap, "ROSMAP Gap Analysis")
    ax_c1.text(-0.2, 1.1, "C", transform=ax_c1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Dual Axis ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.3)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_dual(ax, df, title):
        x = np.arange(len(df['Class']))
        ax.bar(x, df['Support'], color='lightgray', alpha=0.8, width=0.6)
        ax.set_ylabel('Support', color='gray', fontsize=8)
        ax2 = ax.twinx()
        ax2.plot(x, df['F1-score'], color='#E74C3C', marker='o', linewidth=2)
        ax2.set_ylabel('F1-Score', color='#E74C3C', fontsize=8)
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dual(ax_d1, df_brca, "BRCA: Vol vs Qual")
    plot_dual(ax_d2, df_rosmap, "ROSMAP: Vol vs Qual")
    ax_d1.text(-0.25, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.25, hspace=0.3)
    plt.savefig('figure_3_hybrid.png', dpi=300, bbox_inches='tight')
    plt.close()


# ==========================================
# 4. FIGURE 4: DIAGNOSTIC
# ==========================================
def generate_figure_4():
    print("Generating Figure 4: Diagnostic...")
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1])

    # --- Panel A: Rose ---
    gs_a = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 0], wspace=0.25)
    ax_a1 = fig.add_subplot(gs_a[0], polar=True)
    ax_a2 = fig.add_subplot(gs_a[1], polar=True)
    
    def plot_rose(ax, df, title):
        N = len(df); theta = np.linspace(0.0, 2 * np.pi, N, endpoint=False); width = (2*np.pi) / N
        radii = df['F1-score']; colors = plt.cm.viridis(radii / 1.0)
        ax.bar(theta, radii, width=width, bottom=0.0, color=colors, alpha=0.8, edgecolor='white')
        ax.set_xticks(theta); ax.set_xticklabels([f"C{c}" for c in df['Class']], fontsize=9, fontweight='bold')
        ax.set_yticklabels([]); ax.set_ylim(0, 1); ax.set_title(title, fontsize=10, fontweight='bold', pad=20)

    plot_rose(ax_a1, df_brca, "BRCA: Performance Profile")
    plot_rose(ax_a2, df_rosmap, "ROSMAP: Performance Profile")
    ax_a1.text(-0.15, 1.1, "A", transform=ax_a1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel B: Deviation ---
    gs_b = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[0, 1], wspace=0.25)
    ax_b1 = fig.add_subplot(gs_b[0])
    ax_b2 = fig.add_subplot(gs_b[1])

    def plot_dev(ax, df, avg, title):
        x = np.arange(len(df)); dev = df['F1-score'] - avg
        cols = ['#E74C3C' if d < 0 else '#2ECC71' for d in dev]
        ax.bar(x, dev, color=cols, alpha=0.8); ax.axhline(0, color='black', linewidth=1)
        for i, v in enumerate(dev):
            ax.text(i, v + (0.02 if v >=0 else -0.05), f"{v:+.2f}", ha='center', fontsize=8, fontweight='bold')
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_ylim(-0.8, 0.8); ax.set_title(title, fontsize=10, fontweight='bold')

    plot_dev(ax_b1, df_brca, brca_w_avg, "BRCA: Deviation")
    plot_dev(ax_b2, df_rosmap, rosmap_w_avg, "ROSMAP: Deviation")
    ax_b1.text(-0.15, 1.1, "B", transform=ax_b1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel C: Zones ---
    ax_c = fig.add_subplot(gs[1, 0])
    ax_c.add_patch(patches.Rectangle((0, 0), 1, 1, color='#FDEDEC'))
    ax_c.add_patch(patches.Rectangle((0.6, 0.6), 0.4, 0.4, color='#EAFAF1'))
    ax_c.scatter(df_brca['Recall'], df_brca['Precision'], s=df_brca['Support']*8, c='#3498DB', label='BRCA', edgecolors='black')
    ax_c.scatter(df_rosmap['Recall'], df_rosmap['Precision'], s=df_rosmap['Support']*8, c='#E67E22', label='ROSMAP', edgecolors='black')
    for i, txt in enumerate(df_brca['Class']):
        ax_c.annotate(f"B{txt}", (df_brca['Recall'][i], df_brca['Precision'][i]), xytext=(0, -10), textcoords='offset points', ha='center', fontsize=8)
    for i, txt in enumerate(df_rosmap['Class']):
        ax_c.annotate(f"R{txt}", (df_rosmap['Recall'][i], df_rosmap['Precision'][i]), xytext=(0, 5), textcoords='offset points', ha='center', fontsize=8)
    ax_c.set_xlim(0, 1.05); ax_c.set_ylim(0, 1.05); ax_c.set_xlabel("Recall"); ax_c.set_ylabel("Precision")
    ax_c.set_title("Performance Assessment Zones", fontweight='bold', fontsize=10); ax_c.legend(loc='lower left')
    ax_c.text(-0.1, 1.1, "C", transform=ax_c.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    # --- Panel D: Outcomes ---
    gs_d = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=gs[1, 1], wspace=0.25)
    ax_d1 = fig.add_subplot(gs_d[0])
    ax_d2 = fig.add_subplot(gs_d[1])

    def plot_out(ax, df, title):
        x = np.arange(len(df))
        # Correctly calculate values using df columns inside the function
        corr = df['Recall'] * df['Support']
        err = df['Support'] - corr
        
        # Plotting
        ax.bar(x, corr, color='#2ECC71', label='Correct', width=0.6)
        ax.bar(x, err, bottom=corr, color='#E74C3C', label='Error', width=0.6)
        
        ax.set_xticks(x); ax.set_xticklabels(df['Class']); ax.set_title(title, fontsize=10, fontweight='bold')
        if "BRCA" in title: ax.legend(loc='upper right', fontsize=8)

    plot_out(ax_d1, df_brca, "BRCA Outcomes")
    plot_out(ax_d2, df_rosmap, "ROSMAP Outcomes")
    ax_d1.text(-0.2, 1.1, "D", transform=ax_d1.transAxes, fontsize=18, fontweight='bold', va='top', ha='right')

    plt.suptitle(MAIN_TITLE, fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93, wspace=0.2, hspace=0.3)
    plt.savefig('figure_4_hybrid.png', dpi=300, bbox_inches='tight')
    plt.close()

# ==========================================
# 5. EXECUTE ALL
# ==========================================
if __name__ == "__main__":
    generate_figure_1()
    generate_figure_2()
    generate_figure_3()
    generate_figure_4()
    print("All 4 figures generated successfully with title: " + MAIN_TITLE)
```
